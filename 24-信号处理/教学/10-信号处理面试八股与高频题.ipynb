> 一句话速答版：把 00–09 的核心考点压缩成"背下来就有分"的题库。

## 1. 基础概念速答

| 题 | 速答 |
| --- | --- |
| 信号连续/离散 | 自变量是 t 还是 n；数字信号 = 采样+量化 |
| Nyquist | fs ≥ 2fmax；否则混叠 |
| 混叠频率 | $f_{alias}=\|f-kf_s\|$ 取最小；6kHz@8kHz → 2kHz |
| SE vs SD | 标准误 SE=σ/√n（统计）；与采样无关勿混 |
| LTI 判定 | 线性 + 时不变；输出=输入卷积 h |
| 卷积长度 | Lx+Lh−1；FFT 加速需补零到 2 的幂 |

## 2. 变换与频谱速答

| 题 | 速答 |
| --- | --- |
| DFT 复杂度 | O(N²)；FFT 奇偶分解 O(NlogN) |
| 实序列频谱 | 共轭对称，只画一半 |
| 频率分辨率 | Δf=fs/N；帧越长越细 |
| 频谱泄漏 | 截断=乘矩形窗，旁瓣卷积；加 Hann/Hamming |
| 窗主瓣/旁瓣 | 矩形 2 bin/−13dB；Hann 4 bin/−31dB；Hamming 4 bin/−43dB |
| Parseval | Σ\|x\|²=(1/N)Σ\|X\|² |

## 3. 系统与滤波速答

| 题 | 速答 |
| --- | --- |
| FIR vs IIR | FIR 无反馈天然稳定可线性相位；IIR 阶数省有稳定问题 |
| z 变换作用 | 卷积→乘法；z=e^{jω} 是频响 |
| 稳定判据 | 因果系统极点\|p\|<1 |
| 双线性变换 | s=2/T·(1−z⁻¹)/(1+z⁻¹)；无混叠但频率压缩须预扭曲 |
| Butterworth | 最平坦通带；N≥log10(√(10^{As/10}−1)/√(10^{Ap/10}−1))/log10(ωs/ωp) |
| 自适应滤波 | LMS: w←w+μex；μ<2/λmax；NLMS 归一化 |

## 5. 手算例题速查

1. $[1,2,3]*[1,1]=[1,3,5,3]$ 2. 4 点 DFT $[1,2,3,4]\to[10,-2+2j,-2,-2-2j]$ 3. 6kHz@8kHz 混叠 → 2kHz 4. Butterworth Ap=1/As=40/ratio=2 → N=8, Ωc=1.0881 5. LMS 固定激励误差 $1.5\to0.5625\to0.2109$

## 6. 高频追问

**Q：采样率固定时怎么提高频率分辨率？**

加大帧长 N（Δf=fs/N↓）；Welch 则相反（分段牺牲分辨率换方差）。

**Q：语音增强有哪些经典方法？**

谱减、维纳滤波、MMSE-STSA、子空间；深度时代 DCCRN/FRCRN 等（§16-05 延伸）。

**Q：回声消除评估指标？**

ERLE（回声抑制）、AEC 收敛时间、双讲（double-talk）检测。
