## 7️⃣ 手撕清单（6 题必会）

1. 手写皮尔逊相关系数
2. 手写一元 OLS 闭式解
3. 手写双样本 t 检验（Welch）
4. 手写 MLE（正态/泊松）
5. 手写卡方独立性检验
6. 手写 z 检验（两比例，A/B 实验）

```python
import numpy as np
from scipy import stats as st

# 手撕1: 皮尔逊
def my_pearson(x, y):
    xc = x - x.mean(); yc = y - y.mean()
    return (xc*yc).sum() / np.sqrt((xc**2).sum() * (yc**2).sum())

# 手撕2: 一元OLS
def my_ols(x, y):
    b1 = ((x-x.mean())*(y-y.mean())).sum() / ((x-x.mean())**2).sum()
    b0 = y.mean() - b1*x.mean()
    return b0, b1

# 手撕3: Welch t
def my_welch_t(x, y):
    n1, n2 = len(x), len(y)
    s1, s2 = np.var(x, ddof=1), np.var(y, ddof=1)
    t = (x.mean()-y.mean()) / np.sqrt(s1/n1 + s2/n2)
    df = (s1/n1 + s2/n2)**2 / ((s1/n1)**2/(n1-1) + (s2/n2)**2/(n2-1))
    p = 2 * (1 - st.t.cdf(abs(t), df))
    return t, p

# 手撕4: 泊松 MLE
def my_poisson_mle(x):
    return x.mean()

# 手撕5: 卡方独立性
def my_chi2_indep(table):
    table = np.asarray(table, float)
    row, col = table.sum(1, keepdims=True), table.sum(0, keepdims=True)
    exp = row @ col / table.sum()
    chi2 = ((table-exp)**2/exp).sum()
    df = (len(table)-1) * (table.shape[1]-1)
    p = 1 - st.chi2.cdf(chi2, df)
    return chi2, df, p

# 手撕6: 两比例 z
def my_two_prop_z(p1, n1, p2, n2):
    p = (p1*n1 + p2*n2)/(n1+n2)
    se = np.sqrt(p*(1-p)*(1/n1+1/n2))
    return (p1-p2)/se

# ===== 全部验证 =====
rng = np.random.RandomState(7)
x = rng.normal(0, 1, 200); y = 0.5*x + rng.normal(0, .3, 200)
assert abs(my_pearson(x, y) - st.pearsonr(x, y)[0]) < 1e-9
b0, b1 = my_ols(x, y)
assert abs(b0 - 0.0) < 0.1 and abs(b1 - 0.5) < 0.1
a = rng.normal(5, 1, 40); b = rng.normal(5.6, 1.2, 40)
t, p = my_welch_t(a, b)
assert abs(t - st.ttest_ind(a, b, equal_var=False)[0]) < 1e-9
lam = my_poisson_mle(rng.poisson(3.5, 500))
assert abs(lam - 3.5) < 0.3
ct = np.array([[120, 80], [150, 50]])
c2, df, pc = my_chi2_indep(ct)
assert abs(c2 - st.chi2_contingency(ct, correction=False)[0]) < 1e-9
z12 = my_two_prop_z(0.12, 1000, 0.09, 1000)
print('6 道手撕全部通过 ✅')
print('示例: Welch t=%.3f p=%.4f; 两比例 z=%.3f; χ²=%.3f' % (t, p, z12, c2))
```

## 🧮 手算示例：检验手算模板（t 检验全流程）

> 面试手撕黄金模板——拿到数据先算均值→标准差→标准误→检验统计量→p 值。

**数据**：x=[8, 9, 11, 12, 10]，检验 H₀: μ=10

**① 均值**：$\bar x = 50/5 = 10$
**② 样本方差**：$s^2 = [(8-10)^2+...+0+...]/4 = 10/4 = 2.5,\; s \approx 1.581$
**③ 标准误**：$SE = 1.581/\sqrt5 \approx 0.707$
**④ t**：$t = (10-10)/0.707 = 0,\; df=4$
**⑤ p**：双尾 p=1 → 完全无证据拒绝 H₀

**变式**：若 x̄=12，则 t=(12-10)/0.707≈2.828，p≈0.047<0.05 → 拒绝 H₀。

```python
import numpy as np
from scipy import stats as st
x = np.array([8, 9, 11, 12, 10])
n = len(x)
xbar = x.mean(); s = x.std(ddof=1)
se = s / np.sqrt(n)
t = (xbar - 10) / se
p = 2 * (1 - st.t.cdf(abs(t), n-1))
print('均值 =', xbar, ' s(手算≈1.581) =', round(s, 3), ' SE(手算≈0.707) =', round(se, 3))
print('t =', round(t, 3), ' p =', round(p, 3), '→ 不拒绝 H₀')
t_s, p_s = st.ttest_1samp(x, 10)
assert abs(t - t_s) < 1e-9 and abs(p - p_s) < 1e-9
# 变式
x2 = np.array([10, 11, 13, 14, 12])
t2, p2 = st.ttest_1samp(x2, 10)
print('变式: t=%.3f p=%.4f %s' % (t2, p2, '→ 拒绝 H₀' if p2 < 0.05 else '→ 不拒绝'))
print('✅ 手算模板验证通过')
```
