## 7️⃣ 手撕清单（6 道）

> 现场写代码/手算：1) GC 含量 2) 反向互补 3) NW 全局比对
> 4) 香农熵 5) Bray-Curtis 6) 等位基因频率

```python
# 手撕1: GC 含量
def gc(s): return (s.count('G') + s.count('C')) / len(s)
# 手撕2: 反向互补
COMP = {'A':'T','T':'A','C':'G','G':'C'}
def rc(s): return ''.join(COMP.get(b, 'N') for b in reversed(s.upper()))
# 手撕3: NW 全局比对得分
def nw(s1, s2, mt=2, ms=-1, gp=-2):
    n, m = len(s1), len(s2)
    dp = [[0]*(m+1) for _ in range(n+1)]
    for i in range(1, n+1): dp[i][0] = dp[i-1][0] + gp
    for j in range(1, m+1): dp[0][j] = dp[0][j-1] + gp
    for i in range(1, n+1):
        for j in range(1, m+1):
            dp[i][j] = max(dp[i-1][j-1] + (mt if s1[i-1]==s2[j-1] else ms),
                           dp[i-1][j] + gp, dp[i][j-1] + gp)
    return dp[n][m]
# 手撕4: Shannon 熵
import numpy as np
def shannon(counts):
    p = np.asarray(counts, float); p = p[p > 0] / p.sum()
    return -(p * np.log(p)).sum()
# 手撕5: Bray-Curtis
def bc(x, y): return np.abs(np.asarray(x,float)-np.asarray(y,float)).sum() / (np.asarray(x,float)+np.asarray(y,float)).sum()
# 手撕6: 等位基因频率
def af(hom_ref, het, hom_alt):
    n = hom_ref + het + hom_alt
    return (2*hom_alt + het) / (2*n)

# ===== 验证 =====
assert gc('AAGGCC') == 4/6
assert rc('ATCGGCAT') == 'ATGCCGAT'
assert nw('AG', 'AC') == 1          # A-A +2, G-C -1
assert abs(shannon([25, 15, 10]) - 1.02965) < 1e-4   # 手算 08 节
assert abs(bc([10,4,8,0,2], [3,0,12,6,0]) - 23/45) < 1e-9
assert af(12, 6, 2) == 0.25
print('6 道手撕全部通过 ✅')
print('示例: GC=%.3f NW(AG,AC)=%d Shannon=%.3f AF=%.2f'
      % (gc('AAGGCC'), nw('AG','AC'), shannon([25,15,10]), af(12,6,2)))
```
