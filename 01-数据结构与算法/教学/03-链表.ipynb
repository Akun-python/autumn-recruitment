## §1 心智模型与复杂度

### 链表 vs 数组

| 操作 | 数组 | 链表 |
|------|------|------|
| 随机访问第 $i$ 个 | $O(1)$ | $O(n)$（必须从头走） |
| 头部插入/删除 | $O(n)$（搬移） | $O(1)$ |
| 已知节点后插入 | $O(n)$（搬移） | $O(1)$ |
| 尾部插入 | $O(1)$ 均摊 | 有尾指针 $O(1)$，否则 $O(n)$ |
| 按值查找 | $O(n)$ | $O(n)$ |
| 缓存局部性 | 好 | 差（节点分散） |

### 三大心法

- **哑节点（dummy head）**：统一「空表」与「表头」的处理，避免大量 `if` 分支
- **画图**：每个指针操作前先画两三个节点的图，移动顺序错了就断链
- **快慢指针**：用 `fast` 每次 2 步、`slow` 每次 1 步的「速度差」解决中点、判环、找入环点

## §2 反转链表（LeetCode 206）：迭代 + 递归

**题目**：反转整个链表。

### 迭代版公式（三指针）

设 $prev$ 为已反转段的新头方向、$cur$ 为当前节点、$nxt$ 为后继：

$$\text{loop:}\quad nxt \leftarrow cur.next,\quad cur.next \leftarrow prev,\quad prev \leftarrow cur,\quad cur \leftarrow nxt$$

**关键**：先保存 `nxt` 再改写 `cur.next`，否则链断。

### 递归版公式

$$f(head) = \begin{cases} head & head \text{ 为空或只剩一个节点} \\ f(head.next) \to \text{再接回} \ head & \text{否则} \end{cases}$$

即先反转后继子链，再让 `head.next.next = head`、`head.next = None`。

**复杂度**：时间 $O(n)$，空间迭代 $O(1)$ / 递归 $O(n)$（栈深）。

> 💡 反转链表的所有变体（反转区间、K 组翻转、回文链表）都建立在这 4 行迭代上。

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def build(arr):
    dummy = ListNode()
    p = dummy
    for x in arr:
        p.next = ListNode(x)
        p = p.next
    return dummy.next

def to_list(head):
    res = []
    while head:
        res.append(head.val)
        head = head.next
    return res

def reverse_list(head):
    prev = None
    cur = head
    while cur:
        nxt = cur.next      # 1. 先保存后继
        cur.next = prev     # 2. 反转指针
        prev = cur          # 3. 前进
        cur = nxt
    return prev

def reverse_list_recursive(head):
    if head is None or head.next is None:
        return head
    new_head = reverse_list_recursive(head.next)
    head.next.next = head   # 后继的 next 指回自己
    head.next = None
    return new_head

print('迭代反转:', to_list(reverse_list(build([1, 2, 3, 4, 5]))))   # 期望 [5,4,3,2,1]
print('递归反转:', to_list(reverse_list_recursive(build([1, 2, 3, 4, 5]))))
print('空表:', to_list(reverse_list(None)))                        # 期望 []
print('单节点:', to_list(reverse_list(build([1]))))                # 期望 [1]
```

## §3 环形链表（LeetCode 141 / 142）：Floyd 判圈推导

### 141 判断是否有环

快指针每次 2 步、慢指针每次 1 步。若存在环，快指针必然追上慢指针（速度差 1，环内距离有限）。

### 142 找入环节点（公式推导）

设：

- 头节点到环入口距离 $a$
- 环入口到相遇点距离 $b$
- 环长度 $L$

相遇时慢指针走了 $a + b$，快指针走了 $2(a+b)$，且快指针比慢指针多走了若干整圈：

$$2(a+b) = (a+b) + kL \implies a + b = kL \implies a = kL - b$$

即 **从头节点出发走 $a$ 步，恰等于从相遇点再走 $kL - b$ 步（取模后即为环入口）**：

$$a \equiv L - b \pmod{L}$$

所以相遇后，让一个指针从头出发、另一个从相遇点出发，都每次走 1 步，**再次相遇处就是环入口**。

**复杂度**：时间 $O(n)$，空间 $O(1)$。

```python
def build_cycle(arr, pos):
    dummy = ListNode()
    p = dummy
    nodes = []
    for x in arr:
        p.next = ListNode(x)
        p = p.next
        nodes.append(p)
    if pos >= 0:
        p.next = nodes[pos]      # 尾节点指向第 pos 个节点，成环
    return dummy.next

def has_cycle(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            return True
    return False

def detect_cycle(head):
    # Floyd：先找相遇点，再同步走找环入口
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            p = head
            while p is not slow:
                p = p.next
                slow = slow.next
            return p.val
    return None

print('有环:', has_cycle(build_cycle([3, 2, 0, -4], 1)))      # 期望 True
print('无环:', has_cycle(build_cycle([1, 2], -1)))            # 期望 False
print('入环节点值:', detect_cycle(build_cycle([3, 2, 0, -4], 1)))  # 期望 2
print('入环节点值:', detect_cycle(build_cycle([1, 2], -1)))       # 期望 None
```

## §4 合并有序链表 / 相交链表

### 题 3：合并两个有序链表（LeetCode 21，递归模板）

**递推公式**：

$$\text{merge}(l_1, l_2) = \begin{cases} l_2 & l_1 \text{ 为空} \\ l_1 & l_2 \text{ 为空} \\ l_1 \to \text{merge}(l_1.next, l_2) & l_1.val < l_2.val \\ l_2 \to \text{merge}(l_1, l_2.next) & \text{否则} \end{cases}$$

**复杂度**：时间 $O(n + m)$，空间 $O(n + m)$ 递归栈（迭代版 $O(1)$）。

### 题 4：相交链表（LeetCode 160，双指针公式）

**思路**：两个指针分别从 A、B 出发：走完自己链表后**接上对方的链表**继续走。

**为什么正确**：

$$\text{总步数}_A = len(A) + len(B) = \text{总步数}_B = len(B) + len(A)$$

两指针走过的总节点数相同 → 若有交点，必然在交点处**同时**到达（交点后路径完全重合）；
若无交点，它们将同时到达各自链表末尾的 `None`。

**复杂度**：时间 $O(n + m)$，空间 $O(1)$。

```python
def merge_two_lists(l1, l2):
    if not l1:
        return l2
    if not l2:
        return l1
    if l1.val < l2.val:
        l1.next = merge_two_lists(l1.next, l2)
        return l1
    else:
        l2.next = merge_two_lists(l1, l2.next)
        return l2

print('合并有序链表:', to_list(merge_two_lists(build([1, 2, 4]), build([1, 3, 4]))))  # [1,1,2,3,4,4]
print('一方为空:', to_list(merge_two_lists(None, build([1]))))                        # [1]

def get_intersection_node(head_a, head_b):
    # 走完自己再走对方：总步数相等，首次相遇即交点
    pa, pb = head_a, head_b
    while pa is not pb:
        pa = pa.next if pa else head_b
        pb = pb.next if pb else head_a
    return pa.val if pa else None

# 构造相交链表：A = 1->2->3->4->5，B = 9->3->4->5，交点值为 3
common = build([3, 4, 5])
a_head = ListNode(1); a_head.next = ListNode(2); a_head.next.next = common
b_head = ListNode(9); b_head.next = common
print('相交节点值:', get_intersection_node(a_head, b_head))   # 期望 3
print('不相交:', get_intersection_node(build([1]), build([2])))  # 期望 None
```

## §5 LRU 缓存（LeetCode 146）：哈希 + 双向链表

**题目**：get / put 都 $O(1)$ 的最近最少使用缓存。

**思路**：

- **哈希表**：键 → 节点，实现 $O(1)$ 查找
- **双向链表**：头部是最新使用、尾部是最久未用，实现 $O(1)$ 移动与删除
- get 命中 → 节点移到头部；put 已存在 → 更新并移头部；超容量 → 淘汰尾部

**为什么必须双向链表**：删除尾部/移动节点都需要访问前驱，单向链表要 $O(n)$ 找前驱。

**复杂度**：get / put 均 $O(1)$，空间 $O(capacity)$。

> 💡 Python 面试直接写 `OrderedDict` 即可（move_to_end + popitem），但要能讲清底层原理。

```python
from collections import OrderedDict

class LRUCache:
    # OrderedDict：move_to_end 标记「最近使用」，popitem(last=False) 淘汰最久未用
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.d = OrderedDict()

    def get(self, key: int) -> int:
        if key not in self.d:
            return -1
        self.d.move_to_end(key)
        return self.d[key]

    def put(self, key: int, value: int) -> None:
        if key in self.d:
            self.d.move_to_end(key)
        self.d[key] = value
        if len(self.d) > self.capacity:
            self.d.popitem(last=False)

cache = LRUCache(2)
cache.put(1, 1)
cache.put(2, 2)
print('get(1) =', cache.get(1))      # 期望 1（1 变为最近使用）
cache.put(3, 3)                      # 淘汰最久未用的 2
print('get(2) =', cache.get(2))      # 期望 -1
print('get(3) =', cache.get(3))      # 期望 3
cache.put(4, 4)                      # 淘汰 1
print('get(1) =', cache.get(1))      # 期望 -1
print('get(4) =', cache.get(4))      # 期望 4
```

### LRU 手写版：哈希表 + 双向链表（面试手撕核心）

**为什么能 O(1)？** 两个数据结构各管一件事：

- **哈希表** `key -> 节点指针`：O(1) 定位某个键对应的链表节点；
- **双向链表**：O(1) 增删节点（链表头 = 最近使用 MRU，链表尾 = 最久未用 LRU）。

**get(key)**：哈希查不到 → 返回 -1；查到了 → 把该节点**移到链表头**（表示刚用过），返回 value。
**put(key, value)**：已存在 → 更新 value 并移到链表头；不存在 → **头插**新节点；插入后容量超限 → **删链表尾**（最久未用）。

**为什么必须双向链表而不是单向？** 删除节点时需要把**它前一个节点的 `next` 指向它的后一个**——单向链表拿不到前驱，删除是 O(n)；双向链表每个节点存 `prev`，O(1) 完成删除。这是面试追问的核心点。

**实现技巧**：用 `head` / `tail` 两个**哨兵节点**，头插、删尾都不必判空，代码更干净。

复杂度：get / put 均 O(1)（平均），空间 O(capacity)。


```python
class DLinkedNode:
    __slots__ = ('key', 'value', 'prev', 'next')
    def __init__(self, key=0, value=0):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None

class LRUCache:
    # 手写版：哈希 O(1) 定位节点 + 双向链表 O(1) 增删
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = {}             # key -> DLinkedNode
        self.head = DLinkedNode()   # 哨兵头（其 next 是最近使用 MRU）
        self.tail = DLinkedNode()   # 哨兵尾（其 prev 是最久未用 LRU）
        self.head.next = self.tail
        self.tail.prev = self.head

    def _remove(self, node):
        """从链表中摘除节点（O(1)：删除必须知道前驱 → 所以用双向链表）"""
        node.prev.next = node.next
        node.next.prev = node.prev

    def _add_to_head(self, node):
        """头插：插到哨兵 head 之后"""
        node.prev = self.head
        node.next = self.head.next
        self.head.next.prev = node
        self.head.next = node

    def _move_to_head(self, node):
        self._remove(node)
        self._add_to_head(node)

    def _pop_tail(self):
        """删除最久未用（哨兵 tail 的前一个）"""
        node = self.tail.prev
        self._remove(node)
        return node

    def get(self, key: int) -> int:
        if key not in self.cache:
            return -1
        node = self.cache[key]
        self._move_to_head(node)    # 命中 → 移到链表头（最近使用）
        return node.value

    def put(self, key: int, value: int) -> None:
        if key in self.cache:
            node = self.cache[key]
            node.value = value      # 已存在：更新值并移到链表头
            self._move_to_head(node)
            return
        node = DLinkedNode(key, value)
        self.cache[key] = node
        self._add_to_head(node)     # 新键 → 头插
        if len(self.cache) > self.capacity:
            last = self._pop_tail()        # 容量满 → 删链表尾
            del self.cache[last.key]

cache = LRUCache(2)
cache.put(1, 1)
cache.put(2, 2)
print('get(1) =', cache.get(1))      # 期望 1（1 移到链表头）
cache.put(3, 3)                      # 容量满，淘汰链表尾的 2
print('get(2) =', cache.get(2))      # 期望 -1
print('get(3) =', cache.get(3))      # 期望 3（3 移到链表头）
cache.put(4, 4)                      # 再次满，淘汰链表尾的 1
print('get(1) =', cache.get(1))      # 期望 -1
print('get(4) =', cache.get(4))      # 期望 4
```

## §6 K 个一组翻转（25）/ 回文链表（234）

### 题 6：K 个一组翻转链表（LeetCode 25）

**思路**：哑节点 + 每轮三段式：

1. **数**：从 `prev` 出发数 k 个节点，不足 k 个直接返回
2. **翻**：用双指针翻转 `[start, tail]` 这一段（内部逻辑同反转链表）
3. **接**：`prev.next = 新头`、`prev = 旧头`（旧头成为新尾），再接下一组

**复杂度**：时间 $O(n)$（每组翻转 $O(k)$，共 $n/k$ 组），空间 $O(1)$。

### 题 7：回文链表（LeetCode 234）

**思路**三步：

1. **快慢指针找中点**（快指针 2 步 / 慢指针 1 步）
2. **反转后半段**
3. **逐节点比较**前半段与反转后的后半段

**复杂度**：时间 $O(n)$，空间 $O(1)$（不计算修改链表的临时空间）。

```python
def reverse_k_group(head, k):
    dummy = ListNode(0)
    dummy.next = head
    prev = dummy
    while True:
        # 1. 数 k 个，不足则返回
        tail = prev
        for _ in range(k):
            tail = tail.next
            if tail is None:
                return dummy.next
        start = prev.next          # 本组第一个节点
        nxt_group = tail.next      # 本组后的第一个节点
        # 2. 反转 [start, tail]
        cur = start
        p = nxt_group
        while cur is not nxt_group:
            nxt = cur.next
            cur.next = p
            p = cur
            cur = nxt
        # 3. 接回：prev.next = 新头(p)，prev 前进到旧头（新尾）
        prev.next = p
        prev = start

print('k=2:', to_list(reverse_k_group(build([1, 2, 3, 4, 5]), 2)))  # 期望 [2,1,4,3,5]
print('k=3:', to_list(reverse_k_group(build([1, 2, 3, 4, 5]), 3)))  # 期望 [3,2,1,4,5]
print('k=1:', to_list(reverse_k_group(build([1, 2, 3]), 1)))        # 期望 [1,2,3]

def is_palindrome(head):
    # 1. 快慢指针找中点
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    # 2. 反转后半段（从 slow 开始）
    prev = None
    while slow:
        nxt = slow.next
        slow.next = prev
        prev = slow
        slow = nxt
    # 3. 比较
    left, right = head, prev
    while right:
        if left.val != right.val:
            return False
        left = left.next
        right = right.next
    return True

print('回文偶数长:', is_palindrome(build([1, 2, 2, 1])))     # 期望 True
print('回文奇数长:', is_palindrome(build([1, 2, 3, 2, 1])))  # 期望 True
print('非回文:', is_palindrome(build([1, 2])))                # 期望 False
```

### 哑节点技巧总结

$$\text{dummy} \to \text{head}$$

需要「返回新头」或「头节点可能被删除/修改」时（删除倒数第 N 个、K 组翻转、合并），用哑节点免除空表特判。

## §8 常见 bug 与边界（面试踩坑区）

- **先保存后修改**：`cur.next = ...` 前必须先存 `nxt`，否则后继丢失
- **空指针解引用**：`fast.next.next` 前必须判 `fast and fast.next`
- **循环条件**：`head` 为 None 的遍历用 `while head:`；判环用 while 快指针
- **递归爆栈**：链表递归反转在 $n$ 很大时栈溢出，面试先写迭代版
- **没有哑节点**：在头部插入/删除时特判写错导致断链
- **成环陷阱**：测试用 `to_list` 前确保链表无环，否则死循环
- **值比较 vs 引用比较**：判环/相交用 `is`（同一对象），不是 `==`（值相等）
- **K 组翻转的不足 k 组**：剩余不足 k 个要保持原序不翻转（题 25 要求）

### 9.2 为什么快慢指针能找中点

## §高频补充：删除链表倒数第 N 个结点（LeetCode 19）/ 两数相加（LeetCode 2）

**19**：哑节点 + 快慢指针——快指针先走 N 步，再与慢指针同步走，快指针到尾时慢指针恰好停在**待删节点的前驱**；哑节点统一处理「删头结点」。
**2**：模拟竖式加法，从低位到高位逐位相加，`carry` 记录进位；**循环条件要带 `carry`**，否则最后一位进位会丢。


```python
def remove_nth_from_end(head, n):
    # 哑节点 + 快慢指针：慢指针最终停在待删节点的前驱
    dummy = ListNode(0, head)     # 哑节点：统一处理删头结点
    fast = slow = dummy
    for _ in range(n):            # 快指针先走 n 步
        fast = fast.next
    while fast.next:              # 快慢同步走，快到尾时 slow 是前驱
        fast = fast.next
        slow = slow.next
    slow.next = slow.next.next    # 跳过待删节点
    return dummy.next

print('删倒数第2个:', to_list(remove_nth_from_end(build([1, 2, 3, 4, 5]), 2)))  # 期望 [1,2,3,5]
print('删头结点:', to_list(remove_nth_from_end(build([1, 2, 3]), 3)))          # 期望 [2,3]
print('单节点删自己:', to_list(remove_nth_from_end(build([1]), 1)))           # 期望 []
print('删最后一个:', to_list(remove_nth_from_end(build([1, 2]), 1)))          # 期望 [1]
```

```python
def add_two_numbers(l1, l2):
    # 链表模拟加法：低位对齐逐位相加，carry 处理进位
    dummy = ListNode(0)
    cur = dummy
    carry = 0
    while l1 or l2 or carry:      # 最后一位进位也不能漏
        s = carry
        if l1:
            s += l1.val
            l1 = l1.next
        if l2:
            s += l2.val
            l2 = l2.next
        cur.next = ListNode(s % 10)
        cur = cur.next
        carry = s // 10
    return dummy.next

print('342+465:', to_list(add_two_numbers(build([2, 4, 3]), build([5, 6, 4]))))        # 期望 [7,0,8]
print('0+0:', to_list(add_two_numbers(build([0]), build([0]))))                       # 期望 [0]
print('9999+1:', to_list(add_two_numbers(build([9, 9, 9, 9]), build([1]))))           # 期望 [0,0,0,0,1] 进位
print('长度不同:', to_list(add_two_numbers(build([1, 8]), build([0]))))               # 期望 [1,8]
```
