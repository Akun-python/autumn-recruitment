## §1 O(n²) 三兄弟：冒泡 / 选择 / 插入

**冒泡**：相邻比较交换，每轮把最大值沉到末尾。**带优化 flag**：某轮无交换即有序提前退出。

**选择**：每轮找剩余最小值和当前位置交换——**交换次数最少（n 次）**，但比较次数固定 $n(n-1)/2$，且**不稳定**（跨距离交换）。

**插入**：像抓扑克牌，把当前元素插入前面有序区。**对近乎有序数组 O(n)**（几乎不移动）——工程常配「插入排序收尾」。稳定。

**交换次数推导**：

$$C_{bubble} = \frac{n(n-1)}{2} \ \text{(比较)} , \quad S_{selection} \le n \ \text{(交换)} , \quad I_{insert} = \sum_{i=1}^{n-1} (\text{逆序对贡献})$$

````python
# ---- matplotlib 中文字体（防图内中文乱码）----
import matplotlib
matplotlib.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'Songti SC', 'Arial Unicode MS', 'DejaVu Sans']
matplotlib.rcParams['axes.unicode_minus'] = False
# -------------------------------------------
# ════════════════════════════════════════════════════════════════════════
# 【单元目标】手写 O(n²) 排序三兄弟：冒泡 / 选择 / 插入排序；
#             用固定种子的随机数组实测各算法「比较次数 / 交换或移动次数」，并用 sorted() 对照正确性。
# 【算法思路】冒泡：相邻元素两两比较，逆序即交换，每轮把当前最大值「冒泡」到数组末尾；
#             选择：每轮在剩余未排序区间里找最小值，与当前位置交换——交换次数最少（最多 n 次）；
#             插入：像抓扑克牌，把当前元素向左插入左侧「已有序」前缀的正确位置。
# 【变量含义】a=待排序数组（函数内先 copy，避免污染外部数据）；n=数组长度；
#             comp=比较次数；swaps/moves=交换（冒泡/选择）/移动（插入）次数；
#             i、j=两层循环指针；cur=插入排序当前待插入元素；mi=选择排序当前轮最小值下标；
#             swapped=冒泡本轮是否发生过交换（提前退出的标志）；trace=True 时返回
#             (排序结果, 比较次数, 交换/移动次数)，否则只返回排序结果。
# 【循环不变量】冒泡：前 i 轮结束后，末尾 i 个元素是全局最大的 i 个且已有序；
#               选择：前 i 轮结束后，位置 [0, i) 是全局最小的 i 个元素且已有序；
#               插入：每轮外循环开始时，位置 [0, i) 的元素已经有序。
# 【关键边界】冒泡内层只需比较到 n-1-i（末尾 i 个已排好，无需再碰）；
#             插入的查找循环条件 j >= 0（不允许越界到 -1）；
#             选择内层从 i+1 扫到 n-1；三算法必须先 copy 数组，trace 模式才统计次数。
# 【复杂度】冒泡 / 选择 / 插入平均与最坏均为 O(n²)，比较次数均为 n(n-1)/2 量级；
#           插入排序对「近乎有序」数组接近 O(n)（几乎不移动）；三者皆原地排序、额外空间 O(1)；
#           稳定性：冒泡稳定、插入稳定、选择不稳定（跨距离交换破坏相等元素顺序）。
# ════════════════════════════════════════════════════════════════════════
# 本单元手写 O(n²) 三兄弟：冒泡 / 选择 / 插入排序，并实测比较与交换次数
# 三个函数接口一致：传入数组 a，trace=True 时返回 (排序结果, 比较次数, 交换/移动次数)
import numpy as np

# 冒泡排序：相邻元素两两比较，逆序就交换，每轮把当前最大值「冒泡」到数组末尾
# 优化点：若某一轮全程没有交换，说明数组已有序，直接提前结束
def bubble_sort(a, trace=False):
    a = a.copy(); n = len(a); comp = 0; swaps = 0   # 复制数组避免污染原数据；comp=比较次数，swaps=交换次数
    for i in range(n - 1):                          # 外层 n-1 轮，每轮确定一个最大值的位置
        swapped = False                             # 记录本轮是否发生过交换（提前退出的标志）
        for j in range(n - 1 - i):                  # 内层只需比较到 n-1-i，末尾 i 个已是排好的最大值
            comp += 1                               # 每次比较都计数
            if a[j] > a[j + 1]:                     # 相邻元素前大后小 → 逆序，需要交换
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1; swapped = True          # 交换次数 +1，并标记本轮有交换
        if not swapped:          # 优化：本轮无交换 → 已有序
            break
    return (a, comp, swaps) if trace else a

# 选择排序：每轮在剩余未排序区间里找最小值，交换到当前最前面的位置
# 特点：交换次数最少（最多 n 次），但比较次数固定 n(n-1)/2；跨距离交换导致不稳定
def selection_sort(a, trace=False):
    a = a.copy(); n = len(a); comp = 0; swaps = 0   # 复制数组；comp=比较次数，swaps=交换次数
    for i in range(n - 1):                          # 外层 n-1 轮：把第 i 小的元素放到位置 i
        mi = i                                      # 先假设位置 i 就是剩余区间里的最小值
        for j in range(i + 1, n):                   # 在 i 后面的区间里逐个扫描更小的值
            comp += 1                               # 每次比较都计数
            if a[j] < a[mi]:
                mi = j                              # 发现更小的值，更新最小值下标
        if mi != i:                                 # 最小值不在原位才交换
            a[i], a[mi] = a[mi], a[i]
            swaps += 1
    return (a, comp, swaps) if trace else a

# 插入排序：像抓扑克牌一样，把当前元素插入到左侧「已有序」部分的正确位置
# 特点：对近乎有序的数组接近 O(n)；用 > 判断相等不移动，因此稳定
def insertion_sort(a, trace=False):
    a = a.copy(); n = len(a); comp = 0; moves = 0   # 复制数组；comp=比较次数，moves=移动次数
    for i in range(1, n):                           # 从第 2 个元素开始，逐个插入左侧有序区
        cur = a[i]; j = i - 1                       # cur 为待插入元素；j 从它左边一位开始往前找
        while j >= 0 and a[j] > cur:   # 向前找插入位
            a[j + 1] = a[j]                         # 比 cur 大的元素整体右移一格
            j -= 1; comp += 1; moves += 1           # 指针左移并统计比较/移动次数
        comp += 1 if j >= 0 else 0                  # 补记 while 条件最后一次失败的比较
        a[j + 1] = cur                              # 把 cur 放进正确位置
    return (a, comp, moves) if trace else a

# —— 三种排序的实测对比 ——
import random
rng = random.Random(0)                              # 固定随机种子：保证每次运行结果可复现
arr = [rng.randint(0, 99) for _ in range(10)]       # 生成 10 个 [0,99) 随机整数作为测试数据
print('原始数组:', arr)
for name, f in [('冒泡', bubble_sort), ('选择', selection_sort), ('插入', insertion_sort)]:
    res, comp, sw = f(arr, trace=True)              # 打开 trace 同时拿到排序结果和统计次数
    print(f'{name}: 比较 {comp} 次, 交换/移动 {sw} 次 → {res}')
print('正确性对照: 与 sorted 一致 =', all(f(arr) == sorted(arr) for f in
      [bubble_sort, selection_sort, insertion_sort]))

print('\n近乎有序数组插入排序优势演示（各类比较次数）:')
near = list(range(100)) + [57]                      # 构造近乎有序数组：0..99 有序，末尾多放一个 57
for name, f in [('冒泡', bubble_sort), ('选择', selection_sort), ('插入', insertion_sort)]:
    _, comp, sw = f(near, trace=True)               # 这里只关心比较次数，排序结果用 _ 丢弃
    print(f'  {name}: 比较 {comp} 次（插入最少：只动了最后 1 个元素）')
````

## §2 希尔排序：插入排序的跳跃版

**思想**：按间隔 `gap` 分组做插入排序（**预排序**让数组大致有序），逐步缩小 gap 到 1（此时是普通插入，但已接近有序——总代价低）。

**步长序列**：经典 $\lfloor n/2 \rfloor, \lfloor n/4 \rfloor, \dots, 1$——最坏 $O(n^2)$；Hibbard 序列 $2^k-1$ 可到 $O(n^{3/2})$。

**不稳定**：gap > 1 时相同元素可能跨组交换顺序。

````python
# ════════════════════════════════════════════════════════════════════════
# 【单元目标】手写希尔排序（带每轮 gap 的中间状态 trace）与归并排序（带分/合过程 trace），
#             并实测中间过程与正确性。
# 【算法思路】希尔：把数组按间隔 gap 分成若干组，组内做插入排序（预排序），再逐步缩小 gap 到 1，
#             最后一步等价于普通插入排序——但此时数组已接近有序，总代价低。
#             归并：分治框架 = 递归拆半 + 合并两个有序数组；用 <= 保证稳定性（相等时先取左半）。
# 【变量含义】a=待排序数组（函数内先 copy）；n=数组长度；
#             gap=当前步长（从 n//2 起每次减半）；passes=每轮 gap 排序后的数组快照列表；
#             cur=当前待插入元素；j=组内向前跳 gap 步的指针；
#             mid=中点下标；L/R=递归排序后的左/右半；i、j=合并时的双指针；out=合并结果列表；
#             depth=递归深度（只用于 trace 打印缩进）。
# 【循环不变量 / 递归终止】希尔：每个 gap 处理完后，任意同组（下标差为 gap 的倍数）内元素有序；
#                                归并：合并时 L[0:i] 与 R[0:j] 已按序放入 out，且 out 前缀有序；
#                                递归终止条件：len(a) <= 1 时数组天然有序，直接返回副本。
# 【关键边界】希尔组内查找条件 j >= gap（避免越界）；gap 循环 while gap > 0，最后一次 gap=1；
#             归并合并循环 while i < len(L) and j < len(R)，结束后必须 extend 剩余半段；
#             稳定性依赖合并用 <= 而非 <；trace 时注意 a[:mid] / a[mid:] 的切分边界。
# 【复杂度】希尔：平均约 O(n^1.3)（视步长序列），最坏 O(n²)，原地 O(1) 额外空间，不稳定；
#           归并：时间恒为 O(n log n)，空间 O(n)（合并需要额外数组），稳定。
# ════════════════════════════════════════════════════════════════════════
# 希尔排序：把数组按间隔 gap 分成多组，各组内做插入排序（预排序），再逐步缩小 gap 到 1
# 思想：先让相距较远的元素大体有序，最后一步才做普通插入排序——此时数组已接近有序，总代价低
def shell_sort(a, trace=False):
    a = a.copy(); n = len(a)                        # 复制数组，不污染原数据
    gap = n // 2; passes = []                       # 步长从 n//2 开始；passes 记录每轮中间状态供 trace 输出
    while gap > 0:                                  # 步长逐步减半，直到 1（最后一次就是标准插入排序）
        for i in range(gap, n):          # 对每个元素在其 gap 组内插入
            cur = a[i]; j = i                       # cur 为当前要插入的元素
            while j >= gap and a[j - gap] > cur:    # 在同组内向前跳 gap 步找插入位置
                a[j] = a[j - gap]                   # 组内比 cur 大的元素整体后移 gap 步
                j -= gap
            a[j] = cur                              # 把 cur 放回同组内的正确位置
        passes.append((gap, a.copy()))              # 记录本轮 gap 排序后的数组快照
        gap //= 2
    return (a, passes) if trace else a

# —— 希尔排序实测：观察每轮 gap 后的中间状态 ——
rng = random.Random(1)                              # 换一个随机种子，生成新测试数据
arr = [rng.randint(0, 99) for _ in range(12)]
print('原始:', arr)
res, passes = shell_sort(arr, trace=True)           # 打开 trace，拿回每轮中间状态
print('每轮 gap 后的中间状态（希尔=分组预排序 → 逐步有序）:')
for gap, state in passes:
    print(f'  gap={gap}: {state}')
print('结果正确:', res == sorted(arr))

# 归并排序：分治 + 合并（稳定），核心框架 = 递归拆半 + 合并两个有序数组
def merge_sort(a, depth=0, trace=False):
    if len(a) <= 1:                                 # 递归出口：长度 0 或 1 的数组天然有序
        return a.copy()
    mid = len(a) // 2                               # 取中点，把数组分成左右两半
    if trace:                                       # trace 模式打印「分」的过程
        print('  ' * depth + f'分: {a}')
    L = merge_sort(a[:mid], depth + 1, trace)       # 递归排序左半部分
    R = merge_sort(a[mid:], depth + 1, trace)       # 递归排序右半部分
    # 合并两个有序数组
    i = j = 0; out = []                             # 双指针 i、j 分别指向左右半段，out 存合并结果
    while i < len(L) and j < len(R):                # 两边都没取完时，每次取更小的那个
        if L[i] <= R[j]:      # <= 保证稳定（相等先取左）
            out.append(L[i]); i += 1                # 左更小（或相等）→ 取左，指针右移
        else:
            out.append(R[j]); j += 1                # 右更小 → 取右，指针右移
    out.extend(L[i:]); out.extend(R[j:])            # 把某一半剩余的元素全部接到末尾
    if trace:                                       # trace 模式打印「合」的结果
        print('  ' * depth + f'合: {out}')
    return out

# —— 归并排序分治过程实测（trace 输出「分/合」中间状态） ——
print('\n归并排序分治过程（中间输出）:')
arr2 = [rng.randint(0, 99) for _ in range(8)]      # 生成 8 个随机数，规模小方便看分治过程
print('原始:', arr2)
res2 = merge_sort(arr2, trace=True)                # 打开 trace，逐层打印分与合
print('结果正确:', res2 == sorted(arr2))
````

## §3 快速排序：分治之王的推导

**partition**（Lomuto，单指针）：选 pivot，小于 pivot 的换到左边，返回 pivot 最终位置：

$$
\begin{aligned}
&i = \text{smallest index of } >\text{pivot region}\\
&\text{for } j \text{ in range}: \quad \text{if } a[j] < pivot: \text{ swap}(a[i], a[j]), i += 1\\
&\text{swap}(a[i], a[hi]) \Rightarrow \text{pivot 就位}
\end{aligned}
$$

**复杂度推导**：每层 partition 扫描 $O(n)$，期望分两半 → 递归深度 $O(\log n)$ → $O(n\log n)$；若每次只分掉 1 个（已排序数组 + 固定 pivot）→ 深度 $O(n)$ → $O(n^2)$。**随机化 pivot 把最坏变成概率事件**。

**不稳定**：partition 的跨区间交换会打乱相等元素顺序。

````python
# ════════════════════════════════════════════════════════════════════════
# 【单元目标】手写 Lomuto partition + 递归快速排序 + 随机化快排，
#             并用 trace 观察每轮 pivot 就位后的数组，量化「已排序数组 + 固定 pivot」的最坏退化。
# 【算法思路】partition（Lomuto 单指针）：以区间最后一个元素 a[hi] 为基准 pivot，
#             一趟把小于 pivot 的元素换到左边、大于等于的留在右边，最后把 pivot 换到分界处就位；
#             快排递归地对 pivot 左右两侧分别排序；随机化版把数组分成 < / == / > pivot 三部分递归。
# 【变量含义】a=数组；lo、hi=当前区间左右端点（含两端）；pivot=基准值；
#             i=「大于 pivot 区域」的起点（同时指向下一个小于 pivot 元素的落点）；
#             j=扫描指针（从 lo 扫到 hi-1）；p=partition 返回的 pivot 最终位置；
#             depth=递归深度（仅 trace 打印）；l / m / r=随机化版三路划分结果。
# 【循环不变量 / 递归终止】partition 过程中始终有 a[lo..i) < pivot 且 a[i..j) >= pivot；
#                                递归终止：quick_sort 在 lo >= hi（区间为空或单元素）时返回；
#                                quick_sort_random 在 len(a) <= 1 时返回。
# 【关键边界】partition 的 j 只扫到 hi-1（pivot 本身不参与比较）；
#             partition 返回 p 后，左右递归区间必须为 (lo, p-1) 与 (p+1, hi)——不能包含 p；
#             quick_sort 首调用 hi=None 需初始化为 len(a)-1 并 copy；随机化版须先交换/随机选下标。
# 【复杂度】平均 O(n log n)（期望每层扫描 O(n)、递归深度 O(log n)）；
#           最坏 O(n²)：已排序数组 + 固定 pivot 时每轮只切掉一个元素，递归深度 O(n)（本单元末尾实测）；
#           随机化 pivot 把最坏变成概率极低事件；平均额外空间 O(log n)（递归栈），不稳定。
# ════════════════════════════════════════════════════════════════════════
# partition（Lomuto 单指针法）：以区间最后一个元素 a[hi] 为基准 pivot
# 作用：一趟把小于 pivot 的都换到左边，返回 pivot 最终就位的位置，使左边全 < pivot、右边全 ≥ pivot
def partition(a, lo, hi):
    # Lomuto：以 a[hi] 为 pivot，返回就位位置
    pivot = a[hi]; i = lo                           # i 指向「大于 pivot 区域」的起点（左边界）
    for j in range(lo, hi):                         # j 从 lo 扫描到 hi-1（不含基准本身）
        if a[j] < pivot:                            # 遇到比 pivot 小的元素
            a[i], a[j] = a[j], a[i]                 # 把它换到左边区域，i 右移
            i += 1
    a[i], a[hi] = a[hi], a[i]                       # 最后把 pivot 换到 i 处 → pivot 就位
    return i

# 快速排序：递归地对 pivot 左右两侧分别排序。默认参数 lo/hi/depth 让递归调用无需显式传区间
def quick_sort(a, lo=0, hi=None, depth=0, trace=False):
    if hi is None:                                  # 首次调用 hi 为 None → 初始化区间为整个数组
        hi = len(a) - 1
        a = a.copy()                                # 复制数组，保证函数不修改外部数据
    if lo >= hi:                                    # 区间为空或只剩一个元素 → 天然有序，直接返回
        return a if depth == 0 and lo == 0 else None
    p = partition(a, lo, hi)                        # 一趟 partition，让 pivot 就位并返回其位置 p
    if trace:
        print('  ' * depth + f'pivot={a[p]:2d} 就位, 左右: {a}')
    quick_sort(a, lo, p - 1, depth + 1)             # 递归排序 pivot 左侧（区间不含 p）
    quick_sort(a, p + 1, hi, depth + 1)             # 递归排序 pivot 右侧（区间不含 p）
    return a if depth == 0 else None                # 只在最外层返回完整排序结果

# 随机化快速排序（列表推导版）：随机选 pivot，把数组分成 <pivot / ==pivot / >pivot 三部分再递归
# 意义：固定 pivot 在已排序数组上会退化到 O(n²)；随机化把最坏情况变成概率极低的事件
def quick_sort_random(a):
    # 随机 pivot 版：先随机交换，防最坏
    if len(a) <= 1:                                 # 递归出口：空数组或单元素天然有序
        return a.copy()
    p = random.randrange(len(a))                    # 随机选一个下标作为基准
    pivot = a[p]
    l = [x for x in a if x < pivot]                 # 小于 pivot 的部分
    m = [x for x in a if x == pivot]                # 等于 pivot 的部分（一次全部归位）
    r = [x for x in a if x > pivot]                 # 大于 pivot 的部分
    return quick_sort_random(l) + m + quick_sort_random(r)   # 左右递归后拼回：左 + 中 + 右

# —— 快速排序实测：trace 打印每轮 pivot 就位后的数组 ——
rng = random.Random(2)                              # 固定种子，保证结果可复现
arr3 = [rng.randint(0, 50) for _ in range(9)]       # 生成 9 个 [0,50) 的随机整数
print('原始:', arr3)
print('快速排序每轮 partition 中间状态:')
res3 = quick_sort(arr3, trace=True)                 # 打开 trace，观察每轮 pivot 就位过程
print('结果:', res3, ' 正确:', res3 == sorted(arr3))
print('随机化版一致:', quick_sort_random(arr3) == sorted(arr3))

# 快排最坏情况演示：已排序数组 + 固定 pivot 对比递归深度
# 用与 partition 相同的递归逻辑走一遍，只统计最大递归深度（不真正排序），用于量化最坏/平均情况
import sys
def depth_used(a):
    d = [0]                                         # 用列表装深度值，便于在嵌套函数 rec 中修改
    def rec(lo, hi, dep):                           # 模拟快排递归：只统计深度，不做实际排序
        if lo >= hi: return
        d[0] = max(d[0], dep)                       # 记录到达过的最大递归深度
        p = partition(a, lo, hi)                    # 调用同一套 partition 逻辑确定切分点
        rec(lo, p - 1, dep + 1); rec(p + 1, hi, dep + 1)
    rec(0, len(a) - 1, 0)
    return d[0]
# 对比结论：固定 pivot 遇已排序数组，递归深度 ≈ n（退化 O(n²)）；随机数据深度 ≈ log n（O(n log n)）
print('\n最坏情况（已排序数组 n=200）: 递归深度 =', depth_used(list(range(200))),
      '（≈ n → O(n²) 时间）')
print('随机数组: 递归深度 =', depth_used([rng.randint(0, 99) for _ in range(200)]),
      '（≈ log n → O(n log n)）')
````

## §4 堆排序：二叉堆手写

**堆**：完全二叉树，父 ≥ 子（大顶堆）。

**两个核心操作**：

- **sift_down（下沉）**：把小的节点往下换——`O(log n)`
- **build_heap（建堆）**：从最后一个非叶子 `n//2-1` 往前逐个下沉——`O(n)`（数学期望，不是 O(n log n)！）

**排序流程**：建堆 → 反复「堆顶与末尾交换 + 缩小范围下沉」，末尾从后往前形成有序区。

**复杂度**：建堆 $O(n)$ + n 次下沉 $O(n\log n)$ → $O(n\log n)$；**不稳定**、原地 $O(1)$ 空间。

````python
# ════════════════════════════════════════════════════════════════════════
# 【单元目标】手写二叉堆三件套：sift_down（下沉）、build_heap（建堆）、heap_sort（堆排序），
#             并计时验证「建堆 O(n)、完整堆排 O(n log n)」的复杂度结论。
# 【算法思路】利用完全二叉树的数组存储：下标 i 的左孩子为 2i+1、右孩子为 2i+2；
#             下沉：把不满足「父 ≥ 子」的节点与较大的孩子反复交换直到堆性质成立；
#             建堆：从最后一个非叶子节点 n//2-1 起从下往上逐个下沉（O(n) 而非 O(n log n)）；
#             排序：反复把堆顶（当前最大值）与堆末尾交换并缩小堆范围，末尾从后往前形成有序区。
# 【变量含义】a=数组（原地操作）；i=当前下沉的节点下标；n=当前堆的大小（随排序推进而缩小）；
#             l / r=左右孩子下标；big=两个孩子中较大者的下标；
#             end=当前堆的右边界（end 之后是已排好的有序区）；trace=True 时打印中间状态。
# 【循环不变量】sift_down 结束后，以 i 为根的子树满足「父 ≥ 子」的大顶堆性质；
#               heap_sort 每轮开始时：[0, end) 是合法大顶堆，[end, n) 是全局最大的 n-end 个元素（有序）。
# 【关键边界】下沉循环条件 2*i+1 < n（无左孩子即无孩子，直接停止）；
#             右孩子越界（r >= n）时只能选左孩子；
#             建堆必须从 n//2-1 倒序到 0（从下往上），否则上层交换会破坏已建好的子树；
#             堆顶与末尾交换后，下沉范围是 [0, end) 而非 [0, n)，避免触碰已排好的有序区。
# 【复杂度】建堆 O(n)（数学期望求和可证）；排序阶段 n 次下沉各 O(log n)，合计 O(n log n)；
#           全程原地，额外空间 O(1)；因堆顶与末尾的远距离交换，堆排序不稳定。
# ════════════════════════════════════════════════════════════════════════
# 大顶堆下沉（sift_down）：把节点 i 与较大的孩子反复交换，直到满足「父 ≥ 子」
# n 表示当前堆的大小（堆排序时随有序区扩大而缩小），是防止越界的关键参数
def sift_down(a, i, n):
    # 大顶堆下沉：a[i] 与较大的孩子交换
    while 2 * i + 1 < n:             # 有左孩子
        l, r = 2 * i + 1, 2 * i + 2                  # 左孩子下标 l、右孩子下标 r
        big = l if r >= n or a[l] >= a[r] else r     # 选出较大的孩子（右孩子越界或左 ≥ 右时选左）
        if a[i] >= a[big]:                           # 父节点已不小于较大孩子 → 堆性质满足，停止
            break
        a[i], a[big] = a[big], a[i]                  # 父与较大孩子交换
        i = big                                      # 继续向下检查被换下去的原父节点

# 建堆：从最后一个非叶子节点 n//2-1 开始，从下往上逐个下沉，整体复杂度 O(n)
# 必须从下往上（先保证子树各自成堆），不能从根往下——否则根的下沉会被子树里的交换破坏
def build_heap(a):
    for i in range(len(a) // 2 - 1, -1, -1):   # 从最后一个非叶开始
        sift_down(a, i, len(a))                      # 对每个节点下沉，恢复堆性质

# 堆排序流程：建堆 → 反复把堆顶（当前最大值）与末尾交换并缩小堆范围 → 末尾从后往前形成有序区
# 复杂度：建堆 O(n) + n 次下沉 O(n log n) = O(n log n)；原地 O(1) 额外空间，但交换跳远 → 不稳定
def heap_sort(a, trace=False):
    a = a.copy(); n = len(a)                        # 复制数组，不污染原数据
    build_heap(a)                                   # 第一步：把数组原地建成大顶堆
    if trace:
        print('建堆后:', a)
    for end in range(n - 1, 0, -1):                 # end 从最后一个位置往前收缩，表示当前堆的边界
        a[0], a[end] = a[end], a[0]     # 堆顶(最大)→有序区
        sift_down(a, 0, end)            # 缩小堆范围
        if trace and end > n - 4:                   # trace 模式打印前几轮交换后的状态
            print(f'  end={end}: 无序区下沉... 数组={a}')
    return a

# —— 堆排序实测：trace 打印建堆结果与逐轮交换过程 ——
rng = random.Random(3)                              # 固定种子，保证结果可复现
arr4 = [rng.randint(0, 99) for _ in range(10)]
print('原始:', arr4)
res4 = heap_sort(arr4, trace=True)                  # 打开 trace，观察建堆与逐轮交换
print('结果:', res4, ' 正确:', res4 == sorted(arr4))

# 建堆复杂度直觉：n=1e5 计时对比（建堆 vs 排序全程）
import time
big = [rng.randint(0, 10**9) for _ in range(100000)]   # 10 万个随机大整数
t0 = time.time(); build_heap(big); t1 = time.time()     # 只建堆，计时验证 O(n)
t2 = time.time(); heap_sort(big); t3 = time.time()      # 完整堆排序，计时验证 O(n log n)
print(f'\nn=100000: 建堆 {t1-t0:.4f}s（O(n)）, 完整堆排 {t3-t2:.4f}s（O(n log n)）')
````

## §5 非比较排序：计数 / 基数 / 桶（线性时间三兄弟）

**打破 $O(n\log n)$ 下界**：比较排序信息论下界 $\log_2(n!) \approx n\log n$；非比较排序用**数值结构**（桶/位）绕过比较。

**计数排序**：值域 $[0, k)$，统计频次 → 前缀和 → 稳定放置：

$$\text{pos}(x) = \text{prefixsum}[x] - 1, \quad \text{prefixsum}[x] = \sum_{v \le x} count[v]$$

**基数排序**：按位（个/十/百…）依次计数排序——`d` 位 × `O(n+k)`，`d` 小则近线性。

**桶排序**：均匀分成 $k$ 个桶，桶内排序（插入），再按序合并——数据均匀时 $O(n)$ 平均。

**适用条件（必背）**：计数-整数小值域；基数-固定位数的数字/定长字符串；桶-均匀分布实数。

````python
# ════════════════════════════════════════════════════════════════════════
# 【单元目标】手写三种「非比较」线性排序：计数排序 / 基数排序 / 桶排序，并实测正确性与性能。
#             它们绕过比较排序的 O(n log n) 信息论下界，借助数值结构（桶 / 位）做到近线性时间。
# 【算法思路】计数：统计频次 → 求前缀和 → 倒序遍历放置（倒序是稳定性的关键）；
#             基数：从最低位到最高位逐位做稳定计数排序（d 位 × O(n + base)）；
#             桶：把数据按值均匀分到 k 个桶，桶内各自排序（此处复用插入排序），再按桶顺序合并。
# 【变量含义】a=输入数组；K=计数排序值域（数据须在 [0, K) 内）；
#             cnt=频次数组；pref=前缀和数组（长度 K+1，便于取 pref[x+1]）；out=输出数组；
#             maxv=基数排序最大值（决定位数）；exp=当前位权（1=个位，10=十位…）；base=进制；
#             buckets=桶排序的桶列表；k=桶的个数；x=当前元素。
# 【循环不变量】计数排序倒序遍历放置时，每放一个 x 都取该值区间的末端位置，等值元素保持原相对顺序；
#               基数排序每处理完一位（exp），数组在该位及更低位上已稳定有序；
#               桶排序合并阶段按桶下标从小到大输出，桶内有序 ⇒ 整体有序。
# 【关键边界】计数排序要求所有值 ∈ [0, K)，负数 / 超值域需先平移或换算法；
#             基数排序位数循环条件 while maxv // exp > 0（否则最高位排不到就提前结束）；
#             桶排序分桶用 int(x * k) % k（把 [0,1) 切成 k 段），并对每个桶内元素做插入排序；
#             归并/计数/基数的稳定性依赖「倒序放置」或「<= 取左」，写反即丢失稳定。
# 【复杂度】计数：时间 O(n + K)，空间 O(K)（K 过大内存爆炸）；
#           基数：时间 O(d·(n + base))，空间 O(n + base)；桶：均匀分布时平均 O(n)，数据聚集时退化 O(n²)。
# ════════════════════════════════════════════════════════════════════════
# 计数排序：适用于值域已知且较小的整数，统计频次 → 前缀和 → 倒序放置，复杂度 O(n + K)
# 倒序放置是关键：同值元素按原顺序从后往前放，保证稳定性（也是基数排序按位排序的根基）
def counting_sort(a, K=100, trace=False):
    # 值域 [0, K)
    cnt = [0] * K                                   # cnt[i] 统计数值 i 出现的次数
    for x in a:
        cnt[x] += 1                                 # 第一遍扫描：统计每个值的频次
    pref = [0] * (K + 1)                            # pref 为前缀和数组（长度 K+1，方便取 pref[x+1]）
    for i in range(K):
        pref[i + 1] = pref[i] + cnt[i]     # 前缀和
    out = [0] * len(a)                              # 结果数组
    for x in reversed(a):                  # 倒序遍历保证稳定
        pref[x + 1] -= 1                            # 为 x 分配一个位置（下标从对应区间末端往前取）
        out[pref[x + 1]] = x                        # 把 x 放到该位置，等值元素保持原相对顺序
    if trace:
        print('频次 cnt[:12] =', cnt[:12])
        print('前缀和 pref[:13] =', pref[:13])
    return out

# 基数排序：从最低位到最高位，按位做计数排序；d 位 × O(n + base)，d 小则接近线性时间
# 每轮对 exp 对应的一位做稳定排序（借助 counting 的倒序放置），所有位处理完即整体有序
def radix_sort(a, base=10):
    if not a:
        return a                                    # 空数组直接返回
    a = a.copy()                                    # 复制数组，不污染原数据
    maxv = max(a)                                   # 最大值决定需要处理多少位
    exp = 1                                         # exp = 10^当前位（1=个位，10=十位……）
    while maxv // exp > 0:                 # 按每一位计数排序
        cnt = [0] * base                            # 计数数组，统计这一位上各数字的频次
        for x in a:
            cnt[(x // exp) % base] += 1             # 取出 x 在当前位上的数字并计数
        for i in range(1, base):
            cnt[i] += cnt[i - 1]                    # 前缀和：得到每个数字区间的末尾位置
        out = [0] * len(a)
        for x in reversed(a):                       # 倒序放置 → 保持本位的稳定性
            d = (x // exp) % base                   # 当前位的数字
            cnt[d] -= 1                             # 该数字区间的可用位置左移一位
            out[cnt[d]] = x                         # 放到对应位置
        a, exp = out, exp * base                    # 更新数组，并进位到下一位
    return a

# 桶排序：把数据按值均匀分到 k 个桶，桶内各自排序（这里用插入排序），最后按桶顺序合并
# 适用于 [0,1) 均匀分布的实数；若数据分布均匀，每个桶很小，整体接近 O(n)
def bucket_sort(a, k=5):
    # 用于均匀分布的 [0,1) 实数
    buckets = [[] for _ in range(k)]                # 创建 k 个空桶
    for x in a:
        buckets[int(x * k) % k].append(x)           # 按 x*k 的整数部分分桶（把 [0,1) 切成 k 段）
    out = []
    for b in buckets:
        b = insertion_sort(b)                  # 桶内插入排序（必须接收返回值！）
        out.extend(b)                               # 按桶的顺序依次合并到结果
    return out

# —— 非比较排序三兄弟实测：计数 / 基数 / 桶 ——
rng = random.Random(4)                              # 固定种子，保证结果可复现
arr5 = [rng.randint(0, 20) for _ in range(12)]      # 值域 [0,20] 的随机整数，适合计数排序
print('原始:', arr5)
res5 = counting_sort(arr5, K=21, trace=True)        # 打开 trace 看频次与前缀和的中间结果
print('计数排序结果:', res5, ' 正确:', res5 == sorted(arr5))

big_n = [rng.randint(0, 999) for _ in range(15)]    # 三位数随机整数，适合基数排序演示
print('\n基数排序:', radix_sort(big_n), ' 正确:', radix_sort(big_n) == sorted(big_n))

uni = [rng.random() for _ in range(12)]             # 生成 [0,1) 均匀分布的实数
resb = bucket_sort(uni, k=4)                        # 分成 4 个桶排序
print('桶排序结果（[0,1) 均匀分布）:', [round(x, 3) for x in resb])
print('桶排序正确:', resb == sorted(uni))

import math
n = 100000
print('\n大数据实测（n=100000 随机整数）:')
big2 = [rng.randint(0, 10**6) for _ in range(n)]    # 10 万个 0~100万 的随机整数
for name, f in [('内置sorted', lambda x: sorted(x)),
                ('归并', merge_sort),
                ('计数', lambda x: counting_sort(x, K=10**6+1)),
                ('基数', radix_sort)]:
    t0 = time.time(); f(big2.copy()); t1 = time.time()   # 各组独立计时（传副本避免互相影响）
    print(f'  {name:12s} {t1-t0:.3f}s（计数/基数在线性区，值域不大时更有优势）')
````

## §6 稳定性证明与「为什么稳定重要」

| 算法 | 稳定性论证 | 一句话理由 |
|------|-----------|-----------|
| 冒泡 | 相等时不交换 | 相邻交换可控 |
| 插入 | 相等元素插在已有元素**后面** | 从右向左找，`>` 才移动 |
| 归并 | 合并取左（L[i] <= R[j]） | 分治合并可控 |
| 选择 | ❌ 跨距离交换破坏 | 最小元素跳跃换位 |
| 快排 | ❌ partition 跨区间交换 | 跳跃换位 |
| 堆排 | ❌ 堆顶与末尾交换 | 长距离跳跃 |

**为什么工程要稳定**：多关键字排序（先按时间、再按分数——第二次排序保持第一次的顺序）、数据库 ORDER BY 多列、Python `sorted`（Timsort）内建稳定保证。

**稳定实现技巧**：归并合并用 `<=`；计数/基数倒序放回天然稳定；插入用 `>` 不用 `>=`。

````python
# ════════════════════════════════════════════════════════════════════════
# 【单元目标】实现一个通用的排序稳定性检验器 is_stable，并逐一实测 7 种排序算法，
#             与前面理论表（冒泡/插入/归并/计数稳定，选择/快排/堆排不稳定）对照。
# 【算法思路】把每个元素包装成 (值, 原始下标) 对：先用被测算法 f 对纯值序列排序，
#             再按「值序 + 同值按下标升序」重建一个已知稳定的参照序列，
#             最后比较每个值在原始数组与参照结果中的下标顺序是否一致：
#             一致 ⇒ 被测算法稳定；被破坏 ⇒ 不稳定。
# 【变量含义】f=被测排序函数；a=输入数组；K=测试值域上限；
#             paired=(值, 原始下标) 对的列表；vals=纯值序列；sorted_vals=f 的排序结果；
#             by_val=字典（键=值，值=该值出现过的所有原始下标列表）；
#             res=「已知稳定」参照下的排序结果；orig / got=某个值在原始 / 结果中的下标顺序。
# 【循环不变量】参照重建过程中，by_val[v] 的 pop(0) 始终取剩余下标里最小的 ⇒ 同值按下标升序放置。
# 【关键边界】by_val 用 setdefault 收集下标（重复值多个下标要全收集）；
#             用大量重复值（值域只有 0..3）的数组测试更能暴露不稳定；
#             计数排序作被测算法时须传对应 K，否则越界。
# 【复杂度】检验器本身要逐值重建序列，教学演示用；被测算法的复杂度由其自身决定，
#           结论与理论表一致：冒泡/插入/归并/计数稳定，选择/快排/堆排不稳定。
# ════════════════════════════════════════════════════════════════════════
# 稳定性测试器：给每个元素打上原始下标，排序后检查「同值元素的下标顺序」是否保持 → 判断排序是否稳定
def is_stable(f, a, K=10):
    paired = [(v, i) for i, v in enumerate(a)]      # 把数组包装成 (值, 原始下标) 对
    def stable_sort(pairs):                         # 构造一个「已知稳定」的排序参照（值序 + 同值按下标升序）
        vals = [v for v, _ in pairs]                # 取出纯值序列
        sorted_vals = f(vals)                       # 调用被测排序算法 f 对纯值排序
        # 重建对：按值排序，同值用原始下标升序（稳定）
        by_val = {}                                 # 键=值，值=该值出现过的所有原始下标列表
        for v, i in pairs:
            by_val.setdefault(v, []).append(i)      # 收集每个值的全部原始下标
        out = []
        for v in sorted_vals:                       # 按排序后的值序列逐个重建
            out.append((v, by_val[v].pop(0)))       # 同值按下标从小到大取 → 模拟稳定行为
        return out
    # 检查降序下标保持：结果中同值出现顺序与原始下标顺序一致
    res = stable_sort(paired)                       # 得到「已知稳定」参照下的排序结果
    for v, idxs in {}.fromkeys(sorted(set(a)), None).items():   # 遍历每个不同的值
        orig = [i for vv, i in paired if vv == v]   # 该值在原始数组中的下标顺序
        got = [i for vv, i in res if vv == v]       # 该值在排序结果中的下标顺序
        if orig != got:                             # 顺序不一致 → 相对顺序被破坏 → 不稳定
            return False
    return True

# —— 稳定性实测：用大量重复值数组逐一对各排序算法做稳定性检验 ——
test_arr = [rng.randint(0, 3) for _ in range(50)]   # 大量重复值
print('稳定性实测（大量重复值的数组）:')
for name, f in [('冒泡', bubble_sort), ('插入', insertion_sort),
                ('归并', merge_sort), ('选择', selection_sort),
                ('快排', quick_sort_random), ('堆排', heap_sort),
                ('计数', lambda x: counting_sort(x, K=4))]:
    print(f'  {name}: {"✅ 稳定" if is_stable(f, test_arr, K=4) else "❌ 不稳定"}')   # 稳定打 ✅，不稳定打 ❌
print('\n结论与理论表一致：O(n²) 中冒泡/插入稳定，选择不稳定；O(n log n) 中归并稳定，快排/堆排不稳定')
````

## §7 常见 bug 与边界（面试踩坑区）

- **快排已排序数组爆栈**：固定 pivot 最坏 O(n²)/深递归 → 随机 pivot 或三数取中
- **partition 区间边界**：`lo > hi` 提前返回；pivot 就位后左右是 `p-1` / `p+1`（不能含 p）
- **堆排下标**：左孩子 `2i+1`、右 `2i+2`；`sift_down` 的 n 是**当前堆大小**（随 end 缩小）
- **建堆方向**：必须从 `n//2-1` 往前下沉（从下往上），不是从 0 往后
- **归并稳定性**：合并用 `<=`（等于取左），写成 `<` 就丢稳定性
- **计数排序值域假定**：负数/大值域先平移或换基数；K 过大内存爆炸
- **基数排序位数循环**：`while maxv // exp > 0` 否则高位没排到
- **桶排序均匀性**：数据聚类严重时一个桶塞满 → 退化 O(n²)

## §8 面试追问与扩展（加分项）

### 8.1 一个数组，要求稳定且 O(n log n) → 归并

快排/堆排都不稳定；归并稳定。但归并空间 O(n)——工程（Timsort）用**小段插入 + 归并**兼顾两者。

### 8.2 Python sorted 用什么（Timsort）

Timsort = 归并 + 插入混合：识别**有序 run**（自然序列），小 run 用插入排序（<64），run 间归并，检测已有序提前完成——实际数据上比纯归并快，且稳定。

### 8.3 Top-K 与排序的关系

找第 K 大：堆（O(n log k)）或**快选 QuickSelect**（partition 后单侧递归，期望 O(n)）。

### 8.4 逆序对（归并的隐藏能力）

归并合并时「右元素先出 = 与左侧剩余元素构成逆序」→ 顺手统计逆序对数量，O(n log n)。

````python
# ════════════════════════════════════════════════════════════════════════
# 【单元目标】实现两个排序的「隐藏能力」：QuickSelect 求第 k 小/大（Top-K 标准解法）与
#             归并思想统计逆序对数量，并用 min / max / sorted 交叉验证正确性。
# 【算法思路】QuickSelect：复用 partition 让 pivot 就位，比较 pivot 位置 p 与目标 k：
#             相等即答案；p < k 只递归右半；p > k 只递归左半——每层只处理一侧，期望 O(n)。
#             逆序对：归并合并时，若右段元素 R[j] 先被取走，说明它与左段剩余全部元素构成逆序，
#             顺手累加 cnt += len(L) - i，总复杂度仍为 O(n log n)。
# 【变量含义】a=数组；k=0-based 第 k 小；lo、hi=当前区间端点；p=partition 返回的 pivot 位置；
#             mid=中点；L / R=左右半段的有序结果；c1 / c2=左右半各自的逆序对数；
#             cnt=累计逆序对数（c1 + c2 + 跨半新增）；i、j=合并双指针；out=合并结果；arr7=手工测试数组。
# 【循环不变量 / 递归终止】QuickSelect 每轮递归区间都仍包含第 k 小元素（只排除确定不包含答案的一侧）；
#                                终止条件：p == k 直接返回；count_inversions 在 len(arr) <= 1 时返回 (arr, 0)。
# 【关键边界】p < k 时递归右半 (p+1, hi)，p > k 时递归左半 (lo, p-1)——必须跳过 p；
#             逆序对计数是「右元素先出」时才 cnt += len(L) - i，左元素先出不产生新逆序；
#             归并稳定用 <=；count_inversions 只返回逆序对数（排序结果丢弃）。
# 【复杂度】QuickSelect 期望 O(n)、最坏 O(n²)（与快排同源）；
#           逆序对统计 O(n log n)（合并步骤常数级附加统计，不增加量级）。
# ════════════════════════════════════════════════════════════════════════
# 加分演示：快选 QuickSelect 找第 K 小/大（利用 partition 只递归一侧，期望 O(n)）——Top-K 问题的标准解法
def quick_select(a, k, lo=0, hi=None):
    # k: 0-based 第 k 小
    if hi is None:                                  # 首次调用：区间初始化为整个数组
        hi = len(a) - 1
        a = a.copy()                                # 复制数组，避免修改外部数据
    p = partition(a, lo, hi)                        # 一趟 partition：pivot 就位，p 是它的最终位置
    if p == k:                                      # pivot 恰好落在第 k 位 → 它就是答案
        return a[p]
    if p < k:                                       # 答案在 pivot 右侧（第 k 小更大）→ 只递归右半
        return quick_select(a, k, p + 1, hi)
    return quick_select(a, k, lo, p - 1)            # 答案在 pivot 左侧 → 只递归左半

# 逆序对（归并思想）：合并两个有序段时，若右段元素先被取走，说明它与左段剩余元素全部构成逆序对
# 在归并排序的合并步骤里顺手统计，总复杂度仍是 O(n log n)
def count_inversions(a):
    def rec(arr):                                   # 内部递归函数：返回 (排好序的 arr, 逆序对数)
        if len(arr) <= 1:                           # 递归出口：单元素无逆序对
            return arr.copy(), 0
        mid = len(arr) // 2
        L, c1 = rec(arr[:mid]); R, c2 = rec(arr[mid:])   # 递归左右两半，带回各自的逆序对数
        i = j = 0; out = []; cnt = c1 + c2          # 双指针合并；总数 = 左半 + 右半 + 跨半新增
        while i < len(L) and j < len(R):
            if L[i] <= R[j]:
                out.append(L[i]); i += 1            # 左小或相等 → 取左，不产生新的逆序
            else:
                out.append(R[j]); cnt += len(L) - i   # 右边小 → 与左侧剩余全成逆序
                j += 1
        out.extend(L[i:]); out.extend(R[j:])        # 剩余元素直接拼接
        return out, cnt
    return rec(a.copy())[1]                         # 只取逆序对数，排序结果丢弃

# —— 快选与逆序对实测：用 min/max/sorted 交叉验证正确性 ——
arr6 = [rng.randint(0, 99) for _ in range(12)]
print('快选演示:', arr6)
print('  第 0 小（最小）:', quick_select(arr6, 0), '= min?', quick_select(arr6, 0) == min(arr6))     # 与 min 对照
print('  第 5 小:', quick_select(arr6, 5), '= sorted 第5个?', quick_select(arr6, 5) == sorted(arr6)[5])   # 与 sorted 对照
print('  第 11 小（最大）:', quick_select(arr6, 11), '= max?', quick_select(arr6, 11) == max(arr6))   # 与 max 对照

arr7 = [3, 1, 2]                                    # 手工构造小数组，便于人工核对逆序对
print('\n逆序对 [3,1,2]:', count_inversions(arr7), '（期望 2: (3,1),(3,2)）')
print('逆序对 [1,2,3]:', count_inversions([1, 2, 3]), '（期望 0）')       # 升序数组：0 个逆序对
print('逆序对 [3,2,1]:', count_inversions([3, 2, 1]), '（期望 3）')       # 降序数组：任意两两构成逆序对
````
