## 📊 统计推断 8 问速答（上）

**Q1 CLT？** 独立同分布、方差有限 → 样本均值 $\bar X \sim N(\mu,\sigma^2/n)$，不要求总体正态；n≥30 经验线。

**Q2 方差为什么除 n-1？** 均值由同一样本估计，自由度减 1；否则低估方差（有偏）。

**Q3 p 值怎么定义？** H₀ 为真时观察到「当前或更极端」结果的概率；**不是** H₁ 成立的概率。

**Q4 两类错误？** α=误报（拒绝真 H₀），β=漏报（接受假 H₀）；功效 1-β；两者此消彼长，样本量可同时压低。

**Q5 z vs t？** σ 已知用 z、未知用 t（自由度 n-1）；双样本方差齐用合并 t，不齐用 Welch；n 大趋同。

**Q6 χ² 检验？** $\chi^2=\sum(O-E)^2/E$，$E_{ij}=行和×列和/N$，自由度 $(r-1)(c-1)$；期望频数≥5。

**Q7 ANOVA？** $F=\frac{SS_B/(k-1)}{SS_W/(N-k)}$；显著只说明至少一组不同，事后 Tukey；两样本时 $t^2=F$。

**Q8 样本量？** $n \approx \frac{2(z_{1-\alpha/2}+z_{1-\beta})^2}{d^2}$；三要素 α、功效、效应量。

## 📊 统计推断 8 问速答（下）

**Q9 置信区间？** 95% CI = 重复抽样 95% 覆盖真值的区间；$\bar x \pm z_{1-\alpha/2}\cdot SE$；bootstrap 不依赖正态。

**Q10 显著性 vs 效应量？** p 小只说明「差异不太可能是 0」，**效应量（Cohen's d）+ 业务提升 %** 才是决策依据。

**Q11 多重比较？** Bonferroni $\alpha/m$ 控制 FWER（保守）；**BH-FDR**：$p_{(k)}\le\alpha k/m$ 控制误报比例（主流）。

**Q12 分布选择？** 均值→t/z；比例→二项正态近似（n 大）/Fisher；频数→χ²；右偏→对数变换/伽马。

**Q13 离群值？** 先查数据质量（埋点/异常设备），再稳健统计量（中位数/IQR、Winsorize）、稳健回归（Huber）。

**Q14 相关 vs 回归？** 相关度量线性关联强度（无方向）；回归给出「X 变 1 单位 Y 变多少」，可控制协变量——但要小心混淆。

**Q15 多重共线性？** VIF>10 警惕；症状：系数不稳、符号反转；处理：删特征/正则（岭）/PCA。

**Q16 过拟合怎么防？** 交叉验证、正则 L1/L2、早停、数据增强、集成；业务上要「线上验证」最终裁决。

## 🧪 AB 实验 8 问速答（上）

**Q1 流程？** 定义目标 → 假设 → 指标 → 样本量 → 分流 → 跑 → 检验 → 决策；**实验单元/时长/护栏**别忘。

**Q2 样本量怎么算？** 由基线 p₀、MDE、α、功效定；比例用 $n=\frac{(z_α+z_β)^2(p_1(1-p_1)+p_2(1-p_2))}{MDE^2}$。

**Q3 怎么分流？** 哈希（id+盐）分桶；**分层随机**保特征层内 50/50；盐保证实验间正交。

**Q4 转化率检验？** 两比例 z（$z^2=\chi^2_{(1)}$）；均值用 t + Cohen's d；都要报 CI。

**Q5 AA 检验？** 同策略两组 p 应均匀、误报≈5%；显著=分流/埋点坏了。

**Q6 SRM？** A/B 人数偏离 50/50 超误差=样本量比例失衡 → 分流被污染，先修再跑。

**Q7 怎么提功效？** 加样本、降方差（**CUPED** $Y-\theta(X-\bar X)$ 降 $1-\rho^2$、**GAUC** 按用户加权）、选更稳指标。

**Q8 peeking/早停？** 提前看 p 值膨胀误报；用序贯检验或固定时长；显著性≠稳定，多跑几天看趋势。

## 🧪 AB 实验 8 问速答（下）

**Q9 多个指标一起看？** 20 个指标必有假阳性 → 预注册主指标 + FDR 校正；护栏指标（体验/收入）一票否决。

**Q10 结果不显著但想上？** 看效应方向 + CI 上界：是否仍有利？若 MDE 内不显著 → 可能功效不足 → 延长或降方差。

**Q11 辛普森悖论？** 总体无差异、分层有差异（或反之）→ 混淆变量（如新老用户）→ 分层看+预注册分层。

**Q12 实验污染？** 用户跨实验/社交传播/同设备多账号 → 隔离桶、追踪单元、白名单。

**Q13 Novelty 效应？** 新功能短期兴奋 → 周期拉长，看留存/次周活跃。

**Q14 全球 vs 地区？** 不同市场基线差 → 分层/按市场独立实验，全局显著性可能掩盖局部反转。

**Q15 数据不够？** 降方差 > 加流量 > 降显著水平；也可「准实验 + 匹配」但标注证据级别降级。

**Q16 放量条件？** 显著 + 效应≥MDE + 护栏不坏 + AA/SRM 通过 + 方向可解释。

## 🔗 因果推断 8 问速答

**Q1 相关≠因果？** 选择偏差：$E[Y|1]-E[Y|0] = ATE + \underbrace{E[Y(0)|1]-E[Y(0)|0]}_{偏差}$；混淆变量同时影响 T 和 Y。

**Q2 ATE/ATT？** ATE=整体平均处理效应；ATT=处理组内的效应；都基于反事实缺失。

**Q3 随机化为什么好？** T⊥X → 选择偏差=0 → 朴素差即 ATE（AB 金标准）。

**Q4 倾向得分？** $e(x)=P(T=1|x)$ 压缩协变量；给定 e 后 T⊥X；PSM 最近邻匹配 + 平衡性检查（SMD<0.1）。

**Q5 IPTW？** $w=T/e+(1-T)/(1-e)$，Hajek 归一化更稳；极端 e 截断；加权后平衡。

**Q6 DID？** 双重差分消除**固定组间差+共同时间趋势**；前提平行趋势；等价回归交互项 $G\times P$。

**Q7 RDD？** 断点两侧局部随机化；带宽敏感；需检查断点处协变量连续。

**Q8 IV？** Z 满足相关+排他+独立；2SLS 两阶段；用于处理变量内生（遗漏变量/反向因果）。

````python
import numpy as np
from scipy import stats
np.random.seed(0)

# 手撕①：双样本 t（合并方差） + 效应量
def ttest_two(x, y):
    nx, ny = len(x), len(y)
    sp2 = ((nx-1)*x.var(ddof=1) + (ny-1)*y.var(ddof=1))/(nx+ny-2)
    t = (x.mean()-y.mean())/np.sqrt(sp2*(1/nx+1/ny))
    p = 2*(1 - stats.t.cdf(abs(t), nx+ny-2))
    d = (x.mean()-y.mean())/np.sqrt(sp2)
    return t, p, d

rng = np.random.RandomState(1)
a = rng.normal(5.0, 1.0, 80); b = rng.normal(5.5, 1.2, 80)
t, p, d = ttest_two(a, b)
tt, pt = stats.ttest_ind(a, b)
print('t=%.3f p=%.4f d=%.3f | scipy t=%.3f p=%.4f' % (t, p, d, tt, pt))
assert abs(p-pt) < 1e-9
print('✓ 手撕：双样本 t 检验（一场面试必写）')
````

````python
# 手撕②：两比例 z 检验（等价于 2×2 卡方）
def prop_ztest(x1, n1, x2, n2):
    p1, p2 = x1/n1, x2/n2
    pp = (x1+x2)/(n1+n2)
    z = (p1-p2)/np.sqrt(pp*(1-pp)*(1/n1+1/n2))
    return z, 2*(1 - stats.norm.cdf(abs(z)))

z, p = prop_ztest(132, 1000, 110, 1000)
chi2, pchi, _, _ = stats.chi2_contingency([[132, 868], [110, 890]], correction=False)
print('z = %.3f p = %.4f | χ² = %.3f, z² = %.3f' % (z, p, chi2, z**2))
assert abs(chi2 - z**2) < 1e-6
print('✓ 手撕：两比例 z（AB 实验主菜）')

# 手撕③：DID（双重差分）
rng = np.random.RandomState(2)
n = 400
G = rng.rand(n) < 0.5; P = rng.rand(n) < 0.5
y = 10 + 2*G + 3*P + 5*(G*P) + rng.normal(0, 1, n)
did = (y[G & P].mean()-y[G & ~P].mean()) - (y[~G & P].mean()-y[~G & ~P].mean())
X = np.column_stack([np.ones(n), G, P, G*P])
b, *_ = np.linalg.lstsq(X, y, rcond=None)
print('DID = %.3f | 回归交互项 = %.3f（真实 5.0）' % (did, b[3]))
print('✓ 手撕：DID（观察数据因果的万金油）')
````

````python
# 手撕：bootstrap 差置信区间（不依赖正态假设）
def boot_ci(a, b, iters=1000, seed=0):
    rng = np.random.RandomState(seed)
    na, nb = len(a), len(b)
    d = np.array([b[rng.randint(0, nb, nb)].mean() - a[rng.randint(0, na, na)].mean()
                  for _ in range(iters)])
    return d.mean(), np.percentile(d, 2.5), np.percentile(d, 97.5)

rng = np.random.RandomState(4)
a = rng.binomial(1, 0.12, 500); b = rng.binomial(1, 0.15, 500)
d, lo, hi = boot_ci(a, b)
print('Bootstrap 差异 = %.3f, 95%% CI [%.3f, %.3f]' % (d, lo, hi))
print('✓ 手撕：bootstrap（AB 均值/比例差异的稳健 CI）')
````

## 📉 涨跌归因方法

指标跌了：**先分层，再归因，后验证**：

1. **维度拆解**：渠道/新老/地域/端 → 定位「哪个单元贡献了大部分跌幅」；
2. **贡献分解**：单元变化 × 单元占比加权；
3. **排除口径**：埋点/版本/节假日/统计窗口；
4. **验证因果**：不能归因于运营/活动变化时，用 DID/断点确认。

````python
# 涨跌归因：按 渠道×新老 四单元拆解贡献
share = np.array([[0.30, 0.20], [0.25, 0.25]])      # 各单元用户占比
conv0 = np.array([[0.10, 0.08], [0.06, 0.05]])      # 基期转化率
conv1 = np.array([[0.11, 0.08], [0.065, 0.055]])    # 本期转化率
delta = conv1 - conv0
contrib = delta*share
print('转化率变化矩阵：', np.round(delta, 4))
print('加权贡献矩阵：', np.round(contrib, 4))
print('总提升 = %.4f（渠道A-新用户贡献最大：%.4f）' % (contrib.sum(), contrib[0, 0]))
print('✓ 归因 = 变化 × 占比：先定位单元，再解释为什么')
````

## 🎯 终面场景题模板

- 「DAU 突然跌 10%，怎么办？」→ 先分层归因（维度×时间×版本）→ 查口径/埋点 → 再谈产品原因 → 提出验证方案；
- 「要不要上这个功能？」→ 假设 → 指标（北极星+护栏）→ 样本量 → 风险 → 实验验证；
- 「线上实验没效果？」→ 执行度（真分流了吗/量够吗）→ 污染（SRM）→ 效应量 vs MDE → 长期 vs 短期；
- 「怎么向老板汇报？」→ 结论先行 + 数据支撑 + 风险边界 + 下一步。

````python
# 手撕：BH-FDR 校正（多指标实验必写）
def bh_fdr(pvals, alpha=0.05):
    p = np.asarray(pvals, float); m = len(p)
    order = np.argsort(p)
    below = np.where(p[order] <= alpha*np.arange(1, m+1)/m)[0]
    sig = np.zeros(m, bool)
    if len(below): sig[order[:below.max()+1]] = True
    return sig

rng = np.random.RandomState(6)
pvals = np.concatenate([rng.uniform(0, 0.002, 4), rng.uniform(0.05, 0.5, 26)])
sig = bh_fdr(pvals)
print('BH 判显著 %d/30（真阳 4 个）| FDR = %.2f' % (sig.sum(), sig[4:].sum()/max(sig.sum(), 1)))
print('✓ 手撕：BH（10 行内写完，面试加分项）')
````

````python
# SQL 留存逻辑模拟（Python 里演示 GROUP BY 口径）
import datetime as dt
rng = np.random.RandomState(0)
users = list(range(100))
reg = {u: dt.date(2024, 1, 1) + dt.timedelta(days=int(rng.randint(0, 7))) for u in users}
active = {u: [reg[u] + dt.timedelta(days=int(d))
              for d in np.unique(rng.randint(0, 14, 4))] for u in users}
from collections import Counter
ret = Counter()
for u, regd in reg.items():
    for d in active[u]:
        ret[(regd, (d-regd).days)] += 1
for (regd, k) in sorted(ret):
    if k in (0, 1, 7):
        n_reg = sum(1 for u in users if reg[u] == regd)
        print('%s 注册 %2d 人 → 第 %d 日留存 %d 人 (%.0f%%)'
              % (regd, n_reg, k, ret.get((regd, k), 0), 100*ret.get((regd, k), 0)/n_reg))
print('✓ SQL 口径：同期群(注册日) × 活跃日差 → GROUP BY 两键计数')
print('  即 SELECT reg_date, DATEDIFF(active,reg), COUNT(DISTINCT uid) ... GROUP BY 1,2')
````

- **S**：业务问题一句话（「新客转化率连续 3 周下跌」）；
- **T**：分析目标（定位原因并给出验证方案）；

- **R**：结果+影响（定位到渠道策略变化，回归后转化回升 X%；沉淀指标监控看板）。

关键：**量化**每一步（跌了多少、各因素贡献占比、验证置信度），结尾给「下一步」。

````python
# 现场速算：比例 z 检验（面试手算题随身工具）
def ztest_prop(x1, n1, x2, n2):
    p1, p2 = x1/n1, x2/n2
    p = (x1+x2)/(n1+n2)
    se = np.sqrt(p*(1-p)*(1/n1 + 1/n2))
    return (p1-p2)/se, 2*(1-stats.norm.cdf(abs((p1-p2)/se)))

z, p = ztest_prop(1200, 10000, 1150, 10000)
print('转化 12.0%% vs 11.5%% (n=1万): z=%.2f p=%.3f → 显著?%s'
      % (z, p, '是' if p < 0.05 else '否'))
z2, p2 = ztest_prop(1200, 5000, 1100, 5000)
print('转化 24.0%% vs 22.0%% (n=5千): z=%.2f p=%.3f → 显著?%s'
      % (z2, p2, '是' if p2 < 0.05 else '否'))
print('✓ 口算逻辑：差异/SE；SE≈√(p̄(1-p̄)·2/n)——两种量级都一眼看穿')
````

## 🧮 业务公式速查

$$ \text{留存率}_N = \frac{\text{第 }N\text{ 日仍活跃}}{\text{当日新增}} \qquad \text{ARPU} = \frac{\text{营收}}{\text{活跃用户}} $$

$$ \text{GMV} = \text{客单价} \times \text{单量} \qquad \text{转化漏斗每段} = \frac{\text{进入下一步}}{\text{本步}}\times 100\% $$

$$ \text{DAU} = \text{新客} + \text{回流} + \text{留存} \qquad \text{健康增长} = \text{留存占比 ↑} $$

$$ \text{预测} = \text{趋势} \times \text{季节} \times \text{周期} + \text{误差} $$

报告模板：**先漏斗/看板定位 → 再留存/拆解归因 → 最后 AB/预测给结论**。

## 🧮 业务指标公式（默写版）

**ARPU / ARPPU**：$\text{ARPU}=\dfrac{\text{营收}}{\text{用户数}}$，$\text{ARPPU}=\dfrac{\text{营收}}{\text{付费用户数}}$；$\text{ARPU}=\text{ARPPU}\times\text{付费率}$。

**LTV**：$\text{LTV}=\dfrac{\text{ARPU}_{月}}{\text{月流失率}}$（几何级数：$\sum_{t} ARPU\cdot(1-c)^{t}$）；健康条件 $\text{LTV}>\text{CAC}$。

**留存**：$\text{次日留存}=\dfrac{\text{D1 回来用户}}{\text{D0 新增}}$；Cohort 视角：$\text{第 }k\text{ 日留存率}$ 随时间衰减曲线。

**漏斗**：$\text{转化率}_i=\dfrac{N_{i+1}}{N_i}$；总转化 = 各级连乘。

**北极星 + 护栏**：北极星（增长方向单一指标）配护栏（质量/成本不崩）。

**帕累托/幂律**：Top 20% 用户贡献 ~80% 营收 → 分层策略（高价值/长尾）。

> 记忆链：营收=用户×付费率×ARPPU；LTV 与 CAC 比较定投放；留存定产品粘性。

````python
# ARPU 分解数字示例：营收 = 用户 × 付费率 × ARPPU
users, pay_users, revenue = 10000, 300, 450000.0
arpu = revenue/users; arppu = revenue/pay_users; pct = pay_users/users
print('ARPU = %.2f 元 = ARPPU(%.2f) × 付费率(%.1f%%)' % (arpu, arppu, 100*pct))
print('拆到哪层补哪层：提付费率(拉新质量) vs 提 ARPPU(客单价/复购)')
assert abs(arpu - 45.0) < 1e-9 and abs(arpu - arppu*pct) < 1e-9
print('✓ 指标树第一课：增长率 ≠ 单因子变动，要分解归因')
````

## 🍵 白话开篇：指标 = 业务的“体温计”

公司不知道“最近赚不赚钱”只看收入行不行？不行——
收入可能靠烧钱买量撑起来，明天就崩。**指标 = 把业务状态量化成可追踪的数字**。

**三个问题决定指标**（定义指标前先问）：
1. **给谁看**：老板看北极星（GMV），运营看转化率，风控看退款率；
2. **回答什么**：增长没？健康吗？哪里漏？——不同问题要不同指标；
3. **口径是什么**：新客怎么算？GMV 含退款吗？——**口径不一致 = 开会吵架源头**。

**好指标四性（背下来）**：可比（口径统一）、可操作（能指导动作）、
及时（能周期更新）、抗操纵（刷单不能虚高）。

**一句话**：指标不是越多越好——先定“决定生死的主指标”，再看“拆分它的过程指标”。

## 🍵 白话讲透 Cohort：同一批人，跟着他们看

**问题**：本月新增 10000 人，下月又新增 12000 人——“留存 40%”怎么算才准？
把两批人混一起看，新客的“次月留存”会被老客的“N 月留存”污染。

**Cohort（同期群）分析**：按**首次使用月份**把人分组，**每批单独跟踪**：
```
         第0月  第1月  第2月  第3月
1月新客   1000    400    300    250
2月新客   1200    500    380     -
3月新客    900    380     -      -
```
横着看：**某批人的留存曲线**（用户粘性）；竖着看：**同月份不同批**（拉新质量变化）。

**热力图格式**（本书代码示例）：行=批次，列=月份，颜色=留存率——
一眼看出“哪批用户留不住”（可能渠道质量差）。

**一句话**：Cohort = “把不同时间进来的人分开记账”——避免新老混杂，看清每批的真实留存。

## 🍵 白话讲透漏斗：用户在哪一步“跑光”了？

**漏斗 = 关键路径上的步骤转化率**：
曝光 → 点击 → 加购 → 支付。每步一缩水，形似漏斗。

**怎么读漏斗（三步法）**：
1. 找**最大缩水**的一步（曝光→点击掉 70%？那是流量不精准/首屏差）；
2. 算**步间转化率**（点击→加购 = 商品页体验；加购→支付 = 支付流程/价格）；
3. **归因**：最大缩水处优先优化（改一步 = 全漏斗受益）。

**常见坑**：
- 漏斗步骤太多 → 每步都要埋点，错一步全错；
- 用户路径非单一（搜索/推荐/广告三条路）→ 按渠道分漏斗；
- 只看整体不看分群（新客漏斗与老客完全不同）。

**一句话**：漏斗 = 定位“哪里漏最多”——优化的优先级排序器。

## 🔍 公式拆解 ①：ARPU / ARPPU / 付费率

$$\text{ARPU}=\frac{\text{总收入}}{\text{活跃用户数}},\qquad \text{ARPPU}=\frac{\text{总收入}}{\text{付费用户数}},\qquad \text{付费率}=\frac{\text{付费用户数}}{\text{活跃用户数}}$$

**逐符号解释**：
- **ARPU**（每活跃用户收入）：全体用户平均——反映“整体赚钱效率”；
- **ARPPU**（每付费用户收入）：只算付费的人——反映“付费用户的价值”；
- **付费率**：付费用户占比——反映“变现覆盖面”；
- **三角关系（面试快问）**：ARPU = ARPPU × 付费率（本书 r9 已验证：45 = 120×0.375）。

**怎么用（涨 ARPU 的两条路）**：
- 提高 ARPPU（涨价/多买/升级）；提高付费率（降价/首单/激励）——
  两条路**方向经常冲突**，要同时看三个数才不误判。

**注意**：ARPU 涨也可能是“分母变小”（流失多）——
**永远配着活跃用户数一起看**，否则会误报“增长”。

## 🔍 公式拆解 ②：留存率与衰减

$$R_k=\frac{\text{第 k 月仍活跃的第 0 月新客}}{\text{第 0 月新客}},\qquad R_k\approx R_1\cdot r^{k-1}$$

**逐符号解释**：
- $R_k$：新客第 k 个月的留存率（k=1 是次月留存，最常被报告）；
- 分子分母都**限定同一批 Cohort**（见 Cohort 章节）——不限定就是假留存；
- 衰减模型：$r$ = 每月保留比例（0.6~0.9）——**留存曲线 ≈ 指数衰减**（幂律/重尾更真实）；
- **次月留存是最敏感的早期信号**：低于 30% 说明“来了就走”，拉新是烧钱。

**直觉（为什么要看曲线不看单点）**：
- 单点（次月 40%）无法判断趋势——曲线平坦（粘性强）还是持续下滑（一锤子买卖）；
- 指数衰减的对数图应为直线：**斜率 = 用户粘性**（斜率缓 = 产品上瘾）。

**LTV 的关键输入**：$\text{LTV}=\text{ARPPU}\times\frac{r}{1-r}$（月留存 r 下的总价值）——
**留存决定 LTV 的大头**（r 从 0.5→0.8，LTV 翻 4 倍）。

## 🔍 公式拆解 ③：LTV 与 CAC——增长的经济账

$$\text{LTV}=\text{ARPPU}\times\underbrace{\frac{r}{1-r}}_{\text{平均留存月数}},\qquad \text{CAC}=\frac{\text{获客总花费}}{\text{新客数}},\qquad \text{LTV/CAC}>3\ \text{是健康线}$$

**逐符号解释**：
- LTV（用户生命周期价值）：一个用户一辈子贡献的钱；
- $r/(1-r)$：几何级数的和——月留存 r 时平均活跃 $1+r+r^2+\dots = 1/(1-r)$ 月（扣首月即 r/(1−r)）；
- CAC（获客成本）：买一个用户花多少钱（投放+红包+渠道分成）；
- 比值：**赚回获客成本的倍数**——LTV/CAC<1 是纯烧钱；1~3 观望；>3 加大投放。

**直觉（为什么 3:1）**：CAC 是前置现金、LTV 是慢慢收回——
要覆盖资金时间成本+流失不确定性+风险，行业经验取 3 倍安全垫。

**数字验证（r9 例）**：ARPPU=30/月，r=0.9 → LTV=30×0.9/0.1=270；
若 CAC=90 → LTV/CAC=3 恰好健康线。

**一句话**：LTV 决定“能烧多少钱获客”，CAC 决定“烧得值不值”。

````python
# Cohort 热力图：行=月份批次，列=第 N 月，颜色=留存率
rng = np.random.RandomState(1)
n_batch = 6
base = np.array([1.00, 0.42, 0.31, 0.25, 0.21, 0.18])      # 理想留存曲线
mat = np.zeros((n_batch, n_batch))
for b in range(n_batch):
    noise = 1 + 0.06*rng.randn(n_batch - b)
    mat[b, :n_batch-b] = base[:n_batch-b]*noise
mat[3, 0] = 1.0; mat[3, 1] = 0.24   # 第 4 批明显偏弱（渠道差）→ 一眼看出
plt.figure(figsize=(7.5, 3.4))
plt.imshow(mat, aspect='auto', cmap='YlGnBu', vmin=0, vmax=1)
for b in range(n_batch):
    for m in range(n_batch - b):
        plt.text(m, b, '%.2f' % mat[b, m], ha='center', va='center', fontsize=8)
plt.xticks(range(n_batch), ['当月', '+1月', '+2月', '+3月', '+4月', '+5月'])
plt.yticks(range(n_batch), ['1月', '2月', '3月', '4月', '5月', '6月'])
plt.xlabel('注册后月份'); plt.ylabel('注册批次')
plt.colorbar(label='留存率')
plt.title('Cohort 热力图：横看留存曲线，竖看批次质量', fontsize=10)
plt.tight_layout(); plt.show()
print('✓ 第 4 批次整体偏低 → 指向“该月渠道/活动质量问题” → 归因起点')
````

````python
# 帕累托曲线：头部 20% 用户贡献多少收入？
rng = np.random.RandomState(5)
n_u = 2000
spend = np.sort(rng.exponential(50, n_u))[::-1]     # 消费右偏（现实常态）
cum = np.cumsum(spend)/spend.sum()
share_top20 = cum[int(0.2*n_u)]
print('前 20%% 用户贡献收入占比 = %.1f%%（长尾产品也常 50-80%%）' % (100*share_top20))
plt.figure(figsize=(7, 2.8))
plt.plot(np.linspace(0, 100, n_u), cum*100, lw=1.8, color='#1f4e79')
plt.axhline(share_top20*100, color='#b45309', ls='--', lw=1)
plt.axvline(20, color='#b45309', ls='--', lw=1)
plt.xlabel('用户累计占比 (%)'); plt.ylabel('收入累计占比 (%)')
plt.title('帕累托曲线：识别“少数人撑起大半个盘”', fontsize=10)
plt.tight_layout(); plt.show()
assert 0.4 < share_top20 < 0.9
print('✓ 含义：头部用户维护要单独做（VIP 策略）；优化头部留存 = 直击收入大头')
````

````python
# 漏斗图：曝光→点击→加购→支付，标出最大缩水步
steps = ['曝光', '点击', '加购', '支付']
vals = np.array([100000, 45000, 12000, 9000])
rates = vals[1:]/vals[:-1]
k_bad = np.argmin(rates)
plt.figure(figsize=(7, 3.2))
for i, (s, v) in enumerate(zip(steps, vals)):
    w = v/vals[0]
    col = '#b45309' if i == k_bad+1 else '#1f4e79'
    plt.barh(i, w, color=col, alpha=0.85)
    plt.text(w+0.01, i, '%d (%.1f%%)' % (v, 100*v/vals[0]), va='center', fontsize=9)
    if i < 3:
        plt.text(0.5, i+0.5, '转化率 %.1f%%' % (100*rates[i]), ha='center',
                 fontsize=8, color='#555')
plt.yticks(range(len(steps)), steps)
plt.xlim(0, 1.25); plt.xlabel('相对规模')
plt.title('漏斗：点击→加购缩水最大（仅 26.7%%）→ 优化商品页/价格感知优先', fontsize=9)
plt.gca().invert_yaxis()
plt.tight_layout(); plt.show()
assert k_bad == 1
print('✓ 定位：点击→加购（45%%→26.7%%）最陡 → 下一步分析加购页流失原因（价格/库存/评价）')
````

1. **口径漂移**：去年 GMV 含退款、今年不含 → 同比全是假象——口径变更要留档；

5. **季节/星期因素**：对比必须同比或环比对齐周期（双十一 vs 平日没法比）；
6. **幸存者偏差**：只看付费用户/留用用户分析“为什么成功”——漏掉流失的教训；
7. **指标打架**：一个优化涨了 A 崩了 B（转化率↑ 客单价↓）→ 主指标要先定；
8. **没有假设就挖数**：数据挖出来的“发现”多为噪音——先提假设再验证（p 值红线同 DS00）。

## 🧭 指标体系方法论：OSM 与 AARRR

**OSM（目标-策略-度量）**——自顶向下搭体系：
- **O**bjective：业务目标（提高新客 90 日复购）；
- **S**trategy：策略（首单优惠/会员提醒/商品推荐）；
- **M**etric：度量（复购率、优惠券核销率、推荐曝光转化率）。
每个策略配 1~2 个指标，避免“指标海洋”。

**AARRR（海盗指标）**——按用户生命周期分：
- **A**cquisition 获取（新增/下载）；**A**ctivation 激活（首次关键行为）；
- **R**etention 留存（次月留存）；**R**evenue 收入（ARPU/ARPPU）；
- **R**eferral 传播（K 因子/分享率）。

**组合用法**：
- 做增长 → AARRR 找瓶颈环节（哪一步最弱）；
- 定体系 → OSM 让每个指标有“为什么存在”；
- 日常 → 北极星 + 3~5 个护栏 + 过程指标树（r9 示例）。

**一句话**：方法论不是模板，是“每个指标都要能回答：它服务于哪个目标、对应哪个策略”。
