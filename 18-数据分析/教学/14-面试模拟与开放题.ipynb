## 2️⃣ 开放估算题：费米问题拆解

```python
# 题目：估算「某城市一天的咖啡销量」
# 拆解：人口 × 咖啡人群占比 × 日均杯数
population = 25_000_000      # 某城市常住人口
coffee_ratio = 0.15          # 喝咖啡人群占比（都市白领/学生等，拍脑袋但合理）
cups_per_day = 1.2           # 该人群日均杯数
est = population * coffee_ratio * cups_per_day
print(f'估算日销量 = {population:,} × {coffee_ratio} × {cups_per_day} = {est:,.0f} 杯 ≈ {est/1e4:,.0f} 万杯')

# 验证量级：每杯 300ml，总水量
print(f'对应每日咖啡总量 ≈ {est*0.3/1000:,.0f} 吨')
print('\n➡️ 费米问题要点：拆分到可估算的因子、量级合理即可、别纠结精确值')
```

## 3️⃣ 开放估算题：GMV upside 估算

```python
# 题目：提升 1pp 转化率能带来多少收入增量？
# 月均 UV × 转化率 × 客单价
uv = 5_000_000            # 月 UV
conv = 0.02               # 当前转化率
aov = 150                 # 客单价
gmv_now = uv * conv * aov
gmv_up  = uv * (conv + 0.01) * aov
delta = gmv_up - gmv_now
print(f'当前月 GMV: {gmv_now/1e6:.1f} 百万')
print(f'转化率 +1pp 后: {gmv_up/1e6:.1f} 百万')
print(f'月增量: {delta/1e6:.1f} 百万（+{delta/gmv_now:.0%}）')
assert delta == uv * 0.01 * aov
print('\n✅ 拆解到「流量 × 转化 × 客单价」即可快速估算')
```
