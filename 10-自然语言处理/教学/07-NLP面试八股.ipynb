## 第一组 · 词表示 / 分词 / 基础（1-10）

1. **中文分词有哪几类方法？** 词典法（最大匹配/MMSeg）、统计法（HMM/CRF）、神经网络法（BERT-based）；当前工业界主要用预训练模型做基于字的标注 + 词典后处理
2. **正向最大匹配的缺点？** 贪心局部最优；歧义严重（如「研究生命起源」）；改进：双向匹配取分少者
3. **为什么要去停用词/词干化？** 降维降噪；但预训练时代尽量少用，可能丢信息
4. **One-hot 的缺点？** 维度灾、无语义相似性、稀疏（V 维向量只有 1 个非零）
5. **词袋模型（BOW）的缺点？** 丢词序、丢语义、稀疏高维；「猫追狗」和「狗追猫」向量相同
6. **TF-IDF 的直觉？** 词在文档里越常见（TF）、在全库越稀有（IDF）越重要；IDF = log(N/df)
7. **TF-IDF 的局限？** 仍是统计稀疏表示，无语义泛化；未见词 OOV 无法处理
8. **什么是 OOV？** 未登录词——训练词表外的词；用子词（BPE/WordPiece）缓解
9. **BPE 怎么做？** 从字符开始，反复合并最高频相邻符号对，直到目标词表大小
10. **N-gram 和语言模型的关系？** n-gram 是统计语言模型的具体实现（马尔可夫假设截断）

## 第一组 · 词表示 / 分词 / 基础（续 11-20）

11. **链式法则在语言模型里怎么用？** $P(w_1..w_n) = \prod P(w_i | w_1..w_{i-1})$；n-gram 假设只需最近 n-1 个词
12. **平滑作用一句话？** 把概率质量从高频事件让渡给低频/未见事件，避免 0 概率
13. **加一平滑和加 k 平滑区别？** k 可调；V 大时加一平滑给未见词过多概率
14. **Perplexity 与平均负对数似然的关系？** PPL = exp(平均负对数似然)
15. **PPL 越低越好吗？** 同一语料/切分下是；跨语料不可比
16. **最大匹配歧义例子？** 「研究生命起源」→ 词典法可能切「研究生/命/起源」；解决：双向最大匹配 + 启发式规则
17. **分词误差会传播吗？** 会：切分错 → 词向量/语法分析/生成全受影响；预训练时代常按字/子词切分降低风险
18. **IDF 的两种形式？** log(N/df)（经典）vs log((N+1)/(1+df))+1（sklearn 平滑版：避免除零、文档数少时更稳）
19. **为什么 BPE 能缓解 OOV？** 子词可组合成未见词（low + est）；词表大小由合并次数控制
20. **tf 归一化（L2）的作用？** 消除文档长度偏差，余弦相似度与长度无关

### 第一组 · 高频追问补充（1-20）

- **"TF-IDF 和 BM25 哪个好？"** BM25 是 TF-IDF 的嫡系改良：TF 部分做了饱和处理（$\frac{(k_1+1)tf}{k_1+tf}$）、长度归一化可调，检索排序普遍优于朴素 TF-IDF。答"TF-IDF 是基础，BM25 是工程改良"即可。
- **"中文现在还要分词吗？"** 预训练时代主流是**按字/子词**切分（BERT 用 WordPiece），把分词交给模型；但词典/分词在检索、知识图谱等领域仍常见。答"模型按子词、系统按词"的分工最显格局。
- **"BPE 词表大小怎么定？"** 太小 → 碎片化（一个词被拆太碎）；太大 → 接近整词表、失去子词泛化；典型 30k~50k。可补一句"Byte-level BPE（GPT-2）连空格都建模，彻底解决 OOV"。
- **"OOV 只能靠子词吗？"** 还有 <UNK> 映射、字符级模型、按更小粒度回退；答"层次化兜底"即可。
- **"IDF 平滑为什么加 1？"** 一防 df=0 除零，二让语料小时估计更稳；加分点是直接说出 **sklearn 默认 log((N+1)/(1+df))+1**。

> 答题示范（四段式）：TF-IDF 是词频与稀有度的乘积（结论）→ $tf(t,d)\cdot\log\frac{N}{df}$（公式）→ "越常见越稀有越重要"（直觉）→ 仍是稀疏静态表示，被词向量/预训练替代（局限）。

````python
# ---------- 手撕 1：TF-IDF + 余弦相似度（检索排序） ----------
import numpy as np
from collections import Counter

np.random.seed(0)
docs = ['我 喜欢 学习 自然语言', '他 喜欢 学习 数学', '我 喜欢 深度 学习']

def tfidf(docs):
    N = len(docs)
    terms = sorted({w for d in docs for w in d.split()})
    df = {w: sum(1 for d in docs if w in d.split()) for w in terms}
    vecs = []
    for d in docs:
        cnt = Counter(d.split())
        v = []
        for w in terms:
            tf = cnt[w] / len(d.split())
            idf = np.log((1 + N) / (1 + df[w])) + 1   # sklearn 平滑 IDF
            v.append(tf * idf)
        vecs.append(np.array(v))
    return terms, vecs

terms, vecs = tfidf(docs)
def cos(a, b): return a @ b / ((np.linalg.norm(a) + 1e-9) * (np.linalg.norm(b) + 1e-9))
# 查询「自然语言」与哪篇最像（查询向量直接按词表置 1）
qv = np.zeros(len(terms))
for w in ['自然语言']:
    if w in terms:
        qv[terms.index(w)] = 1
print('查询向量非零维度:', [terms[i] for i in np.nonzero(qv)[0]])
ranks = sorted(range(len(docs)), key=lambda i: -cos(qv, vecs[i]))
print('检索排序（应 doc0 最相关）:', ranks)
assert ranks[0] == 0
assert cos(qv, vecs[0]) > cos(qv, vecs[1])   # 命中文档必须严格更近
print('要点：TF-IDF 把「文档-词」计数转成可比较的向量，余弦衡量语义接近度')
````

### 第二组 · 高频追问补充（21-40）

- **"PPL 和 loss 到底啥关系？"** PPL = exp(平均负对数似然) = exp(交叉熵)。口算：平均 NLL=2.30 → PPL≈10。再补一句本质：**PPL 是"平均概率的几何倒数"**——模型平均每步给词 1/PPL 的概率，PPL=10 意味着平均 10 选 1，面试官会认为你懂本质。
- **"为什么未平滑 PPL 可能是 inf？"** 测试集撞上 0 概率 n-gram → log 0 = -inf → 平均后仍 -inf。所以平滑不只是"锦上添花"，而是**让评测可计算**的前提。
- **"加 k 平滑的 k 怎么选？"** 网格搜索最小化开发集 PPL；k≈0 相当于不平滑，k 太大把概率质量摊得太平。答"用开发集调"即可。
- **"bigram 就够了吗？"** 高阶（trigram）上下文更准但更稀疏；工程上常用**插值 + 回退**混合多个阶数。一句话：**阶数越高越准也越稀疏，平滑负责在两者间找平衡**。
- **"统计 LM 真被淘汰了吗？"** 作为主模型被神经 LM 取代，但 n-gram 仍用于**词典外兜底、ASR/输入法重排、机器翻译的约束解码**——"淘汰的是主角地位，不是全部用途"。

> 追问链示范：PPL 是什么 → PPL 越低越好吗（同语料才可比）→ 跨语料为什么不比（词表/切分/分布不同）→ 那 BLEU 呢（评测视角从概率变成重叠度）——一条链把 21-40 的题目串起来。

````python
# ---------- 手撕 2：n-gram 平滑 + PPL（15 秒背写版） ----------
import numpy as np
from collections import Counter

corpus = ['我 爱 学习', '我 爱 北京', '他 爱 北京']
toks = [s.split() for s in corpus]
V2 = sorted({w for s in toks for w in s})
c1 = Counter(w for s in toks for w in s)
c2 = Counter((s[i], s[i + 1]) for s in toks for i in range(len(s) - 1))
def p_addk(w2, w1, k=1.0):
    return (c2[(w1, w2)] + k) / (c1[w1] + k * len(V2))
def ppl_smooth(sents):
    lp, n = 0.0, 0
    for s in sents:
        for i in range(1, len(s)):
            lp += np.log(p_addk(s[i], s[i - 1])); n += 1
    return float(np.exp(-lp / n))
print('未见组合 P(北京|学习) 平滑后 = %.4f（不为 0）' % p_addk('北京', '学习'))
print('测试句 PPL = %.2f' % ppl_smooth([['我', '爱', '学习']]))
assert p_addk('北京', '学习') > 0
print('背写要点：c+k 除 c 的前词+V·k；PPL = exp(-平均 log P)')
````

### 第三组 · 高频追问补充（41-60）

- **"负采样为什么偏偏是 3/4 次幂？"** 提高低频词被采样概率、抑制高频词"垄断"训练；3/4 是经验常数，实际可调。再补数学直觉：$P(w)\propto count(w)^{3/4}$ 让采样分布比频率分布更"平"，低频词被抽到的机会相对变大。
- **"CBOW 和 Skip-gram 的梯度差在哪？"** CBOW 把上下文向量**平均**后预测中心词（梯度更新一份平均向量）；Skip-gram 用中心词预测每个上下文词（每个上下文独立贡献梯度）。所以 Skip-gram 对低频词的信号保留更好，代价是更慢。
- **"GloVe 和 Word2Vec 本质区别？"** GloVe 直接拟合全局共现矩阵的对数（矩阵分解视角）；Word2Vec 是局部窗口在线采样（预测视角）。**GloVe 更像"共现矩阵的低秩分解"，训练更快收敛也更稳**。
- **"fastText 为什么能处理 OOV？"** 词表示 = 字符 n-gram 向量之和，未见过的词也能拆成见过的字符片段拼出向量。
- **"Word2Vec 和 SVD 有关系吗？"** 有：负采样目标在理论上对应 PMI 矩阵的分解（Levy & Goldberg 2014）——答出"词向量 ≈ PMI 矩阵低秩近似"是明显的加分项。

> 这组最常见的主问题链：Word2Vec 两种模型 → 负采样为什么 → 3/4 次幂为什么 → 训练完向量能干嘛 → 与 GloVe/fastText 对比。把 41-60 按这条链背，比逐题零散记忆省一半功夫。

````python
# ---------- 手撕 3：Skip-gram 负采样**一步**梯度（数值校验） ----------
import numpy as np
rng = np.random.default_rng(0)
V3, D3 = 10, 4
W3 = rng.normal(0, 0.1, (V3, D3)); C3 = rng.normal(0, 0.1, (V3, D3))
w_i, w_o, negs = 2, 5, [1, 7, 3]          # 中心词、正上下文、负样本

def loss(W3, C3):
    lo = -np.log(1 / (1 + np.exp(-C3[w_o] @ W3[w_i])))
    for k in negs:
        lo -= np.log(1 / (1 + np.exp(C3[k] @ W3[w_i])))
    return lo

# 解析梯度：正样本项 d/dx[-log σ(x)] = σ(x)-1；负样本项 d/dx[-log σ(-x)] = σ(x)
dL_dW = np.zeros(D3)
g1 = (1 / (1 + np.exp(-C3[w_o] @ W3[w_i]))) - 1
dL_dW += g1 * C3[w_o]
for k in negs:
    dL_dW += (1 / (1 + np.exp(-C3[k] @ W3[w_i]))) * C3[k]
print('解析梯度 norm: %.6f' % np.linalg.norm(dL_dW))
# 数值梯度（抽查 2 维）
eps = 1e-6
for j in [0, 2]:
    Wp, Wm = W3.copy(), W3.copy()
    Wp[w_i, j] += eps; Wm[w_i, j] -= eps
    g = (loss(Wp, C3) - loss(Wm, C3)) / (2 * eps)
    print('维度 %d: 解析 %.6f vs 数值 %.6f 误差 %.2e' % (j, dL_dW[j], g, abs(g - dL_dW[j])))
    assert abs(g - dL_dW[j]) < 1e-6
print('负采样梯度手撕正确 ✓ | 面试就写这一步更新：W[center] -= lr * dL_dW')
````

### 第四组 · 高频追问补充（61-80）

- **"梯度消失的数学直觉？"** 误差沿时间反传要连乘 $W_{hh}^T$：若谱半径 $\rho(W_{hh})<1$，范数以 $|\lambda_{max}|^T$ 指数衰减，$T=50$ 时早已"传不动"；$\rho>1$ 则爆炸。一句话：**连乘决定生死**。
- **"LSTM 凭什么缓解消失？"** 细胞状态 $c_t=f_t\odot c_{t-1}+i_t\odot\tilde c_t$ 是**加性**更新，误差可以从 $t$ 直接"抄近路"传回 $t-1$（梯度有一条"高速公路"）；遗忘门偏置初始化 ≈1 让这条路默认通畅。
- **"为什么门用 sigmoid、候选用 tanh？"** sigmoid 输出 (0,1) 当"开关"；tanh 输出 (-1,1) 提供有正有负的"新内容"，且 tanh 在 0 附近梯度≈1，训练稳定。
- **"GRU 少在哪儿？"** GRU 把遗忘门 + 输入门合并成"更新门"，少一个门、少一组参数，效果与 LSTM 相当——"省约 25% 参数、换相近效果"。
- **"梯度裁剪为什么只治爆炸？"** 裁剪是"事后刹车"；消失是"结构性病"（信息根本传不到远处），只能靠 LSTM 加性通路、残差连接、合理初始化从结构上解决。

> 数字敏感点：LSTM 参数口算 4×(d·h + h·h + h)；谱半径是 RNN 梯度消失的"判决书"；T 越长，连乘指数衰减越狠——这三点是第四组的数字题高频。

### 第五组 · 高频追问补充（81-100）

- **"多头注意力为什么有效？"** 每头在**不同子空间**上关注不同关系（语法 vs 指代 vs 位置）；拼接 + out_proj 融合。可补一句"单头也能工作，多头是容量与并行性的提升"。
- **"为什么正弦位置编码能外推？"** 正弦/余弦对任意位置都有定义，训练见过的长度之外也能算出位置向量（可学习位置编码没有这个性质）；RoPE 更进一步把位置编码进 Q/K 的相对内积，让内积只依赖相对距离。
- **"Pre-LN 和 Post-LN 谁稳？"** Post-LN（原论文）经典但深层易爆；Pre-LN 梯度更稳、利于堆叠，但表达力略降——**现在大模型普遍 Pre-LN**。
- **"KV cache 的代价？"** 省去历史 K/V 的重复计算（生成每步只算新 token），代价是显存随序列长度线性增长；长上下文要配合分页存储 / KV 量化。
- **"FlashAttention 怎么省显存？"** 分块计算 + 在线 softmax（维护 running max/sum），不物化 $O(T^2)$ 的注意力矩阵，显存从 $O(T^2)$ 降到 $O(T)$。
- **"温度 T 怎么调？"** T 小 → 分布尖锐（接近贪心，确定性高）；T 大 → 更均匀（更多样）。生成任务常用 T≈0.7~1.0，要求确定性的场景 T→0。

> 第五组是"公式 + 数字"重灾区：√d（d=512 → √d≈22.6）、每头维度 64、复杂度 O(T²d)、KV cache 的省/费——这四个数字背下来，第五组基本过关。

````python
# ---------- 手撕 5：注意力矩阵 + 因果掩码 ----------
import numpy as np

T4, D4 = 4, 3
Q4 = np.random.default_rng(0).normal(size=(T4, D4))
K4 = np.random.default_rng(1).normal(size=(T4, D4))
V4 = np.random.default_rng(2).normal(size=(T4, D4))

sc = Q4 @ K4.T / np.sqrt(D4)
mask = np.triu(np.ones((T4, T4), dtype=bool), k=1)
sc_m = np.where(mask, -np.inf, sc)
p = np.exp(sc_m - sc_m.max(axis=-1, keepdims=True))
p = p / p.sum(axis=-1, keepdims=True)
print('因果注意力权重（下三角为 0）:\n', np.round(p, 3))
assert (p[np.triu_indices(T4, 1)] == 0).all()       # 右上必须全 0
assert np.allclose(p.sum(axis=-1), 1.0)              # 每行归一
print('每行和=1 ✓ | 第 2 个 token 只能看自己与其左侧 ✓')
out4 = p @ V4
print('输出形状:', out4.shape, '（T, d）——这就是 decoder 自注意力的本质')
````

### 第六组 · 高频追问补充（101-120）

- **"NER 为什么用实体级指标？"** token 级会把一个 5 字实体的 1 字边界错误记成 1/5 错；实体级整片段对才算对，更贴近业务对"边界正确"的要求。

> 第六组的主线是"预训练 + 对齐 + 评估"三位一体：BERT/GPT 结构对比 → 微调/提示 → RLHF 三步 → 下游指标怎么选，正好对应 101→120 的题目顺序。

````python
# ---------- 手撕 6（新）：BLEU 手写（n-gram 裁剪精确率 + 长度惩罚） ----------
import numpy as np
from collections import Counter

def bleu(cand, refs, max_n=4, weights=None):
    cand = cand.split()
    refs = [r.split() for r in refs]
    c = len(cand)
    # 参考长度：取与候选长度最接近的（并列时取较长的）
    r_len = len(min(refs, key=lambda ref: (abs(len(ref) - c), -len(ref))))
    bp = 1.0 if c > r_len else np.exp(1 - r_len / c)   # 长度惩罚：候选偏短要罚
    precs = []
    for n in range(1, max_n + 1):
        cand_ng = Counter(tuple(cand[i:i + n]) for i in range(len(cand) - n + 1))
        max_ref = Counter()
        for ref in refs:
            ref_ng = Counter(tuple(ref[i:i + n]) for i in range(len(ref) - n + 1))
            for k, v in ref_ng.items():
                max_ref[k] = max(max_ref.get(k, 0), v)  # 跨参考取最大计数
        clipped = sum(min(v, max_ref.get(k, 0)) for k, v in cand_ng.items())
        total = sum(cand_ng.values())
        precs.append(clipped / total if total else 0.0)
    w = weights or [0.25] * max_n
    geo = 0.0 if any(p == 0 for p in precs) else np.exp(sum(wt * np.log(p) for wt, p in zip(w, precs)))
    return bp * geo, precs, bp, r_len

# 例 1：与参考完全一致 -> BLEU = 1.0
s1, pr1, bp1, rl1 = bleu('the cat is on the mat', ['the cat is on the mat', 'there is a cat on the mat'])
print('例1 BLEU = %.4f | 1-4gram 精确率 %s | BP %.2f' % (s1, np.round(pr1, 3), bp1))
assert np.isclose(s1, 1.0)

# 例 2：重复词被裁剪 -> p1 只有 1/4（the 出现 4 次只算 1 次）
s2, pr2, bp2, rl2 = bleu('the the the the', ['the cat'])
print('例2 BLEU = %.4f | p1 = %.3f（裁剪后 1/4）| BP %.2f（候选不短不罚）' % (s2, pr2[0], bp2))
assert np.isclose(pr2[0], 0.25)

# 例 3：候选比参考短 -> 精确率全 1 仍被长度惩罚
s3, pr3, bp3, rl3 = bleu('the cat is on the', ['the cat is on the mat'])
print('例3 BLEU = %.4f | BP = %.4f（候选 5 词 < 参考 6 词）' % (s3, bp3))
assert bp3 < 1.0

# 例 4：bigram 重复被裁剪 -> p2 = 1/3
s4, pr4, bp4, rl4 = bleu('the cat the cat', ['the cat'])
print('例4 p2 = %.3f（the-cat 出现 2 次只算 1 次）' % pr4[1])
assert np.isclose(pr4[1], 1 / 3)

print('要点：BLEU = BP × exp(Σ w_n log p_n)；p_n 用 max-ref 裁剪防重复刷分')
````

````python
# ---------- 手撕 7（新）：多头注意力维度口算 + 逐头 torch 对照 ----------
import numpy as np
import torch
import torch.nn.functional as F

np.random.seed(0); torch.manual_seed(0)

d_model, h = 512, 8
head_dim = d_model // h
assert d_model == h * head_dim                      # 口算：512 = 8 × 64
print('口算：d_model=512, h=8 -> 每头 head_dim = %d ✓' % head_dim)

T = 8
X = torch.randn(T, d_model)
Q = X; K = X; V = X                                # 简化：投影设为单位变换，专注维度与数值验证

# 手动多头：reshape (T, d_model) -> (T, h, head_dim) -> transpose (h, T, head_dim)
Qh = Q.view(T, h, head_dim).transpose(0, 1)        # (h, T, head_dim)
Kh = K.view(T, h, head_dim).transpose(0, 1)
Vh = V.view(T, h, head_dim).transpose(0, 1)
print('切分后 Q 形状:', tuple(Qh.shape), '-> 每头 (T=%d, %d)' % (T, head_dim))

outs = []
for i in range(h):
    s = Qh[i] @ Kh[i].T / np.sqrt(head_dim)        # 缩放点积
    p = torch.softmax(s, dim=-1)
    assert torch.allclose(p.sum(dim=-1), torch.ones(T))
    o = p @ Vh[i]                                  # (T, head_dim)
    ref = F.scaled_dot_product_attention(Qh[i], Kh[i], Vh[i], dropout_p=0.0)
    assert torch.max(torch.abs(o - ref)).item() < 1e-4, i
    outs.append(o)

out = torch.cat(outs, dim=-1)                      # 拼接回 (T, d_model)
print('拼接后形状:', tuple(out.shape), '（T, d_model）-> 再过 out_proj 才是完整 MHA ✓')
assert out.shape == (T, d_model)
print('8 头全部与 torch.nn.functional.scaled_dot_product_attention 一致（<1e-4）✓')
````

````python
# ---------- 手撕 8（新）：混淆矩阵 -> P/R/F1（macro/micro 派生 + sklearn 对照） ----------
import numpy as np
from sklearn.metrics import precision_recall_fscore_support

np.random.seed(0)
n, K = 120, 3
y_true = np.random.randint(0, K, n)
y_pred = np.random.randint(0, K, n)
y_pred[y_true == 2] = 2                            # 人为让类 2 更准（演示类别不均衡）

# 1) 混淆矩阵：CM[i,j] = 真实 i 被预测成 j
CM = np.zeros((K, K), dtype=int)
for t, p in zip(y_true, y_pred):
    CM[t, p] += 1

# 2) 派生 TP / FP / FN
TP = np.array([CM[i, i] for i in range(K)])
FP = CM.sum(axis=0) - TP                          # 每列：预测成 i 但不是 i
FN = CM.sum(axis=1) - TP                          # 每行：真实 i 但没预测成 i
with np.errstate(divide='ignore', invalid='ignore'):
    P = TP / (TP + FP); R = TP / (TP + FN); F1 = 2 * P * R / (P + R)
P = np.nan_to_num(P); R = np.nan_to_num(R); F1 = np.nan_to_num(F1)

# 3) macro：各类平均
macro_p, macro_r, macro_f1 = P.mean(), R.mean(), F1.mean()

# 4) micro：全局汇总（各类 TP/FP/FN 相加）
tp_all, fp_all, fn_all = TP.sum(), FP.sum(), FN.sum()
micro_p = tp_all / (tp_all + fp_all)
micro_r = tp_all / (tp_all + fn_all)
micro_f1 = 2 * micro_p * micro_r / (micro_p + micro_r)
acc = tp_all / n
assert np.isclose(micro_f1, acc)                  # 样本数一致时 micro-F1 = accuracy

# 5) sklearn 对照
sk_p, sk_r, sk_f, _ = precision_recall_fscore_support(y_true, y_pred, average=None)
sk_macro = precision_recall_fscore_support(y_true, y_pred, average='macro')
sk_micro = precision_recall_fscore_support(y_true, y_pred, average='micro')
assert np.allclose(P, sk_p) and np.allclose(F1, sk_f)
assert np.allclose(macro_f1, sk_macro[2]) and np.allclose(micro_f1, sk_micro[2])

print('混淆矩阵:\n', CM)
print('各类 P/R/F1:', np.round(P, 3), np.round(R, 3), np.round(F1, 3))
print('macro-F1 = %.4f | micro-F1 = %.4f | accuracy = %.4f' % (macro_f1, micro_f1, acc))
print('与 sklearn precision_recall_fscore_support 逐位一致 ✓')
print('要点：micro-F1（样本数一致时）= accuracy；macro 是各类平均，类别不均衡时更敏感')
````

### 面试心法：一个主问题 = 一条追问链

八股面试最怕的不是不会，而是**只会背第一层**。以"讲讲注意力机制"为例，标准追问链长这样：

**Q1 主问题**：讲注意力 → 答 softmax(QKᵀ/√d)V + 直觉"按相关性加权汇总"；
**Q2 为什么除 √d** → 点积方差 ~d_k，防 softmax 饱和（背数字：d=512 → √d≈22.6）；
**Q3 为什么多头** → 多子空间关注不同关系，拼接 + out_proj；
**Q4 复杂度多少** → O(T²d)，长序列怎么办 → FlashAttention / 稀疏注意力；
**Q5 位置信息哪来的** → 位置编码 / RoPE，为什么需要（注意力是置换等变的集合操作）；
**Q6 生成时怎么省算力** → KV cache。

看出规律了吗？**每往下挖一层，都是在"上一个答案的缺陷"上提问**：有了公式问数值、有了数值问复杂度、有了复杂度问优化、有了机制问信息注入、有了推理问工程加速。备考时对每个高频主问题（TF-IDF、PPL、负采样、梯度消失、BERT vs GPT、macro/micro……）都预先推两条追问，面试就不会被"接着问"卡住。本页 120 问 + 上面各组的"高频追问补充"，就是为这条链准备的弹药。

## 面试速答（30 秒背诵版，10 连）

- **为什么除 √d？** 点积方差 ≈ d_k，不除会进 softmax 饱和区、梯度消失。
- **为什么多头？** 不同子空间并行关注不同关系；拼接再过 out_proj。
- **因果掩码怎么实现？** 上三角置 -inf（或 mask_fill），softmax 后右侧为 0。
- **LSTM 为什么缓解梯度消失？** 细胞状态 c 加性更新 = 梯度高速公路；遗忘门偏置初始 ≈1 偏向记住。
- **MLM 为什么 80/10/10？** 缓解预训练见 [MASK]、微调不见的 mismatch。
- **负采样 3/4 次幂？** 提高低频词被采样概率，抑制高频词主导。
- **PPL 和 loss？** PPL = exp(平均 NLL) = exp(交叉熵)；同一语料越低越好。
- **KV cache 省什么？** 省历史 K/V 的重复计算，代价是显存。
- **微调学习率为什么小？** 1e-5~5e-5，防止破坏预训练表示。
- **统计 vs 神经 LM？** 统计稀疏不可泛化；神经参数共享、能泛化到未见组合。

## 高频易错点 TOP 12（面试官最爱挖的坑）

1. **TF-IDF 忘平滑**：IDF 直接用 log(N/df) 遇 df=0 会除零；用 log((N+1)/(1+df))+1。
2. **BOW 忘说丢语序**：举「猫追狗 vs 狗追猫」反例。
3. **PPL 跨语料比较**：PPL 只在同一语料/切分下可比。
4. **负采样梯度符号**：正样本 σ(x)-1，负样本 σ(x)；别搞反。
5. **遗忘门偏置忘初始化**：初始 ≈1~2（偏向记住）。
6. **注意力忘除 √d**：只写 softmax(QKᵀ)V 就丢分了。
7. **掩码位置**：因果掩码是**上三角**置 -inf（右上角），不是下三角。
8. **多头拼接后忘 out_proj**：MHA = concat(heads)·W_O，还有 W_O。
9. **BERT/GPT 方向搞反**：BERT = Encoder 双向（MLM）；GPT = Decoder 单向（自回归）。
10. **MLM 比例误解**：15% 是「选多少 token 掩码」，80/10/10 是「被选中 token 内部」的替换方式。
11. **micro 与 macro 混用**：micro-F1 = accuracy（样本数一致时）；macro 是各类平均。
12. **BLEU 忘长度惩罚**：候选比参考短必须 BP < 1，否则重复短句刷分。
