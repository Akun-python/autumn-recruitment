## 第 0 章 机器学习的统一框架

### 0.1 三要素：数据、模型、优化

机器学习压缩成一句话：**给定数据，找一个函数，让它在数据上犯的错（损失）最小**。

| 要素 | 是什么 | 例子 |
|------|--------|------|
| 数据 $D=\{(x_i,y_i)\}$ | 输入特征 $x$ 与输出 $y$ | 身高体重 → 是否患病 |
| 模型 $f(x;\theta)$ | 一族假设函数，参数 $\theta$ | 直线 $w^\top x+b$、树、神经网络 |
| 优化 | 最小化损失 $L(\theta)$，学出 $\theta^*$ | 梯度下降、闭式解、EM |

**分类套路**：算每个类别的概率 $P(y|x)$ 再取最大（最大后验）：$\hat y = \arg\max_y P(y|x)$。
**回归套路**：直接预测数值，损失常用平方误差。

### 0.2 偏差-方差分解（所有模型都在权衡的事）

$$\mathbb E[(y-\hat f(x))^2] = \underbrace{(\mathbb E[\hat f]-f)^2}_{\text{偏差}} + \underbrace{\mathbb E[(\hat f-\mathbb E[\hat f])^2]}_{\text{方差}} + \sigma^2$$

- **偏差大** = 模型太简单，连规律都学不到（欠拟合）；
- **方差大** = 模型太复杂，记住训练集噪声（过拟合）；
- **噪声**不可消除。

> 🍎 打靶类比：偏差 = 弹着点平均偏离靶心；方差 = 弹着点散布范围。简单模型稳定但偏（高偏差），复杂模型平均准但飘（高方差）。

**防过拟合三板斧**：① 加正则（L1/L2）；② 交叉验证调参；③ 更多数据。

**运行环境**：本 notebook 仅需阅读，无代码；其余章节依赖 Python 3.8 + numpy（对照 cell 需 sklearn）。

## 附录 A：公式速查表（考前 10 分钟）

| 模型 | 核心公式 |
|------|---------|
| 线性回归 | $w^* = (X^\top X)^{-1}X^\top y$；岭回归 $(X^\top X+\lambda I)^{-1}X^\top y$ |
| 逻辑回归 | $p=\sigma(w^\top x)=\frac{1}{1+e^{-w^\top x}}$；$\sigma'=\sigma(1-\sigma)$；$\nabla L=X^\top(p-y)$ |
| 朴素贝叶斯 | $\hat y=\arg\max_y P(y)\prod_j P(x_j|y)$；平滑 $\frac{count+\alpha}{N+\alpha K}$ |
| KNN | $d_2=\sqrt{\sum_j (x_j-x'_j)^2}$；$\hat y = $ k 个近邻投票 |
| SVM | 间隔 $\frac{2}{\|w\|}$；对偶 $\max \sum\alpha_i - \frac12\sum\sum\alpha_i\alpha_j y_i y_j K(x_i,x_j)$；$w=\sum\alpha_i y_i x_i$ |
| 决策树 | $H=-\sum p_k\log p_k$；$\text{Gain}=H(D)-H(D|A)$；Gini $=1-\sum p_k^2$ |
| AdaBoost | $\alpha_m=\frac12\ln\frac{1-e_m}{e_m}$；$D_{m+1}\propto D_m e^{-\alpha_m y G_m}$ |
| GBDT | 负梯度 $r_{im}=-\partial L/\partial f$；$F_m=F_{m-1}+\eta T_m$ |
| XGBoost | 叶子权重 $w^*=-\frac{\sum g}{\sum h+\lambda}$ |
| BP | $\delta^L=(a-y)\odot\sigma'(z^L)$；$\delta^l=(W^{l+1\top}\delta^{l+1})\odot\sigma'(z^l)$；$\partial L/\partial W^l=\delta^l a^{l-1\top}$ |
| K-Means | $\min\sum_k\sum_{i\in C_k}\|x_i-\mu_k\|^2$；分配→更新两步交替 |
| PCA | 拉格朗日 → $\Sigma w=\lambda w$；保留比例 $=\sum_{i\le k}\lambda_i/\sum\lambda_i$ |
| HMM | 前向 $\alpha_{t+1}(j)=[\sum_i\alpha_t(i)a_{ij}]b_j(o_{t+1})$；Viterbi $\delta_t(j)=\max_i\delta_{t-1}(i)a_{ij}\cdot b_j(o_t)$ |
| CRF | $P(Y|O)=\frac{\exp(\sum_t\sum_k\lambda_k f_k)}{Z(O)}$；梯度 = 真实计数 − 模型期望 |
| AUC | $P(score_+ > score_-) = \frac{R_+ - n_+(n_++1)/2}{n_+ n_-}$ |

## 附录 B：算法对比总表（对比题背这张）

| 对比 | 一句话结论 |
|------|-----------|
| LR vs SVM | LR 给概率、支持增量训练；SVM 只认支持向量、可核升维。样本少/高维/非线性 → SVM；要概率/在线 → LR |
| L1 vs L2 | L1 菱形约束 → 稀疏（自动选特征）；L2 圆约束 → 权重均匀缩小 |
| 线性回归 vs 逻辑回归 | 一个回归一个分类；一个平方损失一个交叉熵（$w^\top x$ 直出 vs 过 sigmoid） |
| AdaBoost vs GBDT | AdaBoost 放大错样本权重；GBDT 拟合负梯度。AdaBoost 怕噪声，GBDT 更稳 |
| GBDT vs XGBoost | XGBoost = GBDT + 二阶泰勒 + 叶子正则 + 列采样 + 增益剪枝 |
| Bagging vs Boosting | Bagging 并行降方差；Boosting 串行降偏差 |
| 随机森林 vs GBDT | RF 降方差、抗噪；GBDT 精度高、怕噪声、串行慢 |
| K-Means vs GMM | 硬聚类 vs 软聚类（概率归属）；GMM 是 K-Means 的概率版，都是 EM 思想 |
| PCA vs LDA | 无监督最大方差 vs 有监督最大化类间/类内比 |
| HMM vs CRF | 生成式局部归一化 vs 判别式全局归一化；CRF 特征自由、无标注偏置 |
| 决策树 vs KNN | 树：贪心分区可解释需剪枝；KNN：惰性学习、高维灾难 |
| 过拟合通用解法 | 正则 / 剪枝 / 交叉验证调参 / 更多数据 / 集成降方差 |
